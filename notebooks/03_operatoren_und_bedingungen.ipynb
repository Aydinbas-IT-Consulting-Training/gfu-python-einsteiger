## Arithmetische Operatoren

| Operator | Bedeutung | Beispiel | Ergebnis |
| --- | --- | --- | --- |
| `+` `-` `*` | Grundrechenarten | `7 * 6` | `42` |
| `/` | Division, immer `float` | `7 / 2` | `3.5` |
| `//` | ganzzahlige Division | `7 // 2` | `3` |
| `%` | Rest (Modulo) | `7 % 2` | `1` |
| `**` | Potenz | `2 ** 10` | `1024` |

````python
minutes_total = 135
hours = minutes_total // 60
minutes = minutes_total % 60
print(f"{minutes_total} Minuten = {hours} h {minutes} min")

# divmod liefert beides auf einmal; Entpacken (Notebook 01) verteilt die zwei Werte
hours, minutes = divmod(minutes_total, 60)
print(f"{minutes_total} Minuten = {hours} h {minutes} min")
````

## Exkurs: Fallunterscheidung mit `match`

Dieser Abschnitt ist optional; im Kurs überspringen wir ihn.

Vergleichst du **einen Wert** mit mehreren Möglichkeiten, ist `match` übersichtlicher als eine lange `elif`-Kette. Python prüft die `case`-Zweige von oben nach unten, der erste passende gewinnt.

````python
command = "halt"

match command:
    case "start":
        print("Motor läuft")
    case "stop" | "halt":
        print("Motor aus")
    case _:
        print(f"Unbekannter Befehl: {command}")
````

- `|` bedeutet „oder“: Der Zweig passt auf `"stop"` und auf `"halt"`.
- `case _:` passt auf alles, was übrig bleibt, wie `else`.
- Für Java- und C#-Umsteiger: `match` braucht kein `break`. Es läuft immer genau ein Zweig, danach geht es hinter dem `match` weiter.

### Zusatzbedingungen: Guards

Hinter ein Muster darfst du `if` schreiben. Der Zweig passt nur, wenn auch diese Bedingung wahr ist. `case t if …` speichert den Wert dabei unter dem Namen `t`, damit die Bedingung ihn prüfen kann:

````python
temperature = 12

match temperature:
    case t if t > 25:
        print("heiß")
    case t if t > 15:
        print("angenehm")
    case _:
        print("frisch bis kalt")
````

Für reine Größer-/Kleiner-Vergleiche bleibt `if`/`elif` meist lesbarer. Seine Stärke zeigt `match`, wenn ein Wert eine **Struktur** hat, etwa ein Befehl aus mehreren Wörtern.

Muster für Listen: Notebook 06. `match` zusammen mit Aufzählungen (`Enum`) zeigt Notebook 12 (optional).

> **Typischer Fehler:** Ein nackter Name im `case` vergleicht **nicht**, er passt auf alles und speichert den Wert unter diesem Namen. `case limit:` trifft also immer zu, auch wenn es eine Variable `limit` gibt. Vergleiche deshalb mit festen Werten wie `case "stop":` oder mit Namen, die einen Punkt enthalten (`case Color.RED:`, siehe Notebook 12, optional).

## None und Vergleiche: `==` oder `is`?

`None` kennst du aus Notebook 01 als „kein Wert“, etwa wenn eine Angabe fehlt: Der Kunde hat keinen Rabattcode eingegeben.

Die Regel: Werte vergleichst du mit `==`. `None` prüfst du mit `is None` (oder `is not None`).

````python
discount_code = None  # der Kunde hat keinen Rabattcode eingegeben

if discount_code is None:
    print("Bestellung ohne Rabattcode")
````

### Falle: `None` ist falsy

Wie `0`, `""` und `[]` gilt `None` in einer Bedingung als falsch (Notebook 01: `bool(None)` ist `False`). `if not x:` trifft deshalb auch zu, wenn `x` gültig `0` ist. Meinst du „nicht angegeben“, prüfe mit `is None`:

````python
temperature = 0  # 0 °C ist ein gültiger Messwert, None hieße: Sensor hat nichts geliefert

if not temperature:
    print("if not:     kein Messwert")  # falsch, es gibt einen Messwert

if temperature is None:
    print("if is None: kein Messwert")
else:
    print(f"if is None: {temperature} °C")
````

## Zusammenfassung

- `//` und `%` liefern Ganzzahl-Ergebnis und Rest, `divmod` beides auf einmal.
- Vergleiche liefern `bool` und lassen sich verketten: `0 < x < 100`.
- `and`, `or`, `not` verknüpfen Bedingungen.

- Bedingter Ausdruck: `a if bedingung else b`, nur für kurze Fälle.

- `input()` liefert immer `str`, also umwandeln.
- Werte mit `==` vergleichen, `None` mit `is None`; `if not x:` trifft auch bei `0` und `""` zu.
