## Die `for`-Schleife

`for` läuft über alle Elemente einer Folge, etwa die Zeichen eines Strings oder die Einträge einer Liste:

````python
for letter in "Python":
    print(letter)
````

````python
for city in ["Köln", "Berlin", "München"]:
    print(f"Schulung in {city}")
````

> Für Java- und C#-Umsteiger: Python kennt kein `for (int i = 0; i < n; i++)`. Die `for`-Schleife entspricht `foreach`.

## Zahlenfolgen mit `range()`

`range(start, stop, step)` erzeugt Zahlen von `start` bis **ausschließlich** `stop`.

````python
for i in range(5):
    print(i, end=" ")
````

````python
print(list(range(1, 11)))      # 1 bis 10
print(list(range(0, 21, 5)))   # in 5er-Schritten
print(list(range(10, 0, -1)))  # rückwärts
````

## Summen bilden

Ein häufiges Muster: Eine Variable vor der Schleife anlegen und in der Schleife verändern.

````python
total = 0
for number in range(1, 101):
    total += number
print(f"Summe 1 bis 100: {total}")
````

Für einfache Fälle gibt es eingebaute Funktionen:

````python
sum(range(1, 101))
````

## Index und Wert mit `enumerate()`

Brauchst du zusätzlich die Position, nimmst du `enumerate()` statt `range(len(...))`:

````python
for position, name in enumerate(["Ada", "Grace", "Linus"], start=1):
    print(f"{position}. {name}")
````

## Die `while`-Schleife

`while` wiederholt, **solange** eine Bedingung wahr ist. Sinnvoll, wenn die Anzahl der Durchläufe vorher nicht feststeht.

````python
balance = 1000.0
years = 0
while balance < 2000:
    balance *= 1.05
    years += 1
print(f"Nach {years} Jahren hat sich das Guthaben verdoppelt: {balance:.2f} €")
````

> **Achtung:** Ändert sich die Bedingung nie, läuft die Schleife endlos. Abbrechen im Notebook mit dem Stopp-Button (■), im Terminal mit `Strg + C`.

## `break` und `continue`

- `break` beendet die Schleife sofort.
- `continue` springt zum nächsten Durchlauf.

````python
import random

tries = 0
while True:
    roll = random.randint(1, 6)
    tries += 1
    if roll == 6:
        break
print(f"Nach {tries} Würfen eine 6")
````

````python
for number in range(1, 11):
    if number % 3 == 0:
        continue
    print(number, end=" ")
````

## Verschachtelte Schleifen

Eine Schleife in einer Schleife. Die innere läuft bei **jedem** Durchlauf der äußeren komplett durch.

````python
for row in range(1, 6):
    for col in range(1, 6):
        print(f"{row * col:4}", end="")
    print()
````
