## Die `for`-Schleife

`for` läuft über alle Elemente einer Folge, etwa die Zeichen eines Strings oder die Einträge einer Liste:

````python
for letter in "Python":
    print(letter)
````

````python
for city in ["Köln", "Berlin", "München"]:
    print(f"Schulung in {city}")
````

> Für Java- und C#-Umsteiger: Python kennt kein `for (int i = 0; i < n; i++)`. Die `for`-Schleife entspricht `foreach`.

## Zahlenfolgen mit `range()`

`range(start, stop, step)` erzeugt Zahlen von `start` bis **ausschließlich** `stop`.

````python
for i in range(5):
    print(i, end=" ")
````

````python
print(list(range(1, 11)))      # 1 bis 10; list() zeigt alle Zahlen auf einmal
print(list(range(0, 21, 5)))   # in 5er-Schritten
print(list(range(10, 0, -1)))  # rückwärts
````

## Summen bilden

Ein häufiges Muster: Eine Variable vor der Schleife anlegen und in der Schleife verändern.

````python
total = 0
for number in range(1, 101):
    total += number
print(f"Summe 1 bis 100: {total}")
````

## Index und Wert mit `enumerate()`

Brauchst du zusätzlich die Position, nimmst du `enumerate()` statt `range(len(...))`. `enumerate()` liefert in jeder Runde ein Paar aus Position und Wert; `for position, name in ...` entpackt es in zwei Namen (Entpacken, Notebook 01):

````python
for position, name in enumerate(["Ada", "Grace", "Linus"], start=1):
    print(f"{position}. {name}")
````

## Die `while`-Schleife

`while` wiederholt, **solange** eine Bedingung wahr ist. Sinnvoll, wenn die Anzahl der Durchläufe vorher nicht feststeht.

````python
balance = 1000.0
years = 0
while balance < 2000:
    balance *= 1.05
    years += 1
print(f"Nach {years} Jahren hat sich das Guthaben verdoppelt: {balance:.2f} €")
````

## `break` und `continue`

- `break` beendet die Schleife sofort.
- `continue` springt zum nächsten Durchlauf.

````python
import random

tries = 0
while True:
    roll = random.randint(1, 6)
    tries += 1
    if roll == 6:
        break
print(f"Nach {tries} Würfen eine 6")
````

````python
for number in range(1, 11):
    if number % 3 == 0:
        continue
    print(number, end=" ")
````

## `else` an Schleifen

Eine Schleife kann einen `else`-Block haben. Er läuft, wenn die Schleife **regulär zu Ende** gegangen ist, also **ohne `break`**. Das passt genau zu einer typischen Aufgabe: etwas suchen und reagieren, wenn es nicht gefunden wurde.

````python
guests = ["Ada", "Grace", "Linus"]
wanted = "Guido"

for guest in guests:
    if guest == wanted:
        print(f"{wanted} ist eingeladen")
        break
else:  # kein break: die Schleife lief komplett durch
    print(f"{wanted} steht nicht auf der Liste")
````

Ohne `else` bräuchtest du eine zusätzliche Variable wie `found = False`, die du in der Schleife setzt und danach prüfst. `else` spart das.

> **Merkhilfe:** Lies `else` an einer Schleife als „kein `break`“. Der Name ist unglücklich gewählt, selbst erfahrene Entwickler stolpern darüber. Nutze `else` nur zusammen mit `break` und schreib einen kurzen Kommentar dazu.

`while` kennt `else` genauso: Der Block läuft, wenn die Bedingung falsch wird, aber nicht nach einem `break`.

## Verschachtelte Schleifen

Eine Schleife in einer Schleife. Die innere läuft bei **jedem** Durchlauf der äußeren komplett durch.

````python
for row in range(1, 6):
    for col in range(1, 6):
        print(f"{row * col:4}", end="")
    print()
````

## Zusammenfassung

- `for element in folge:` läuft über alle Elemente.
- `range(start, stop, step)`: `stop` ist nicht enthalten.
- `enumerate()` liefert Position und Wert.
- `while bedingung:` wiederholt, solange die Bedingung gilt.
- `break` beendet, `continue` überspringt.
- `else` an einer Schleife läuft nur, wenn es **kein** `break` gab: ideal für „gesucht, aber nicht gefunden“.
- Verschachtelte Schleifen: Die innere läuft bei jedem Durchlauf der äußeren komplett durch.
