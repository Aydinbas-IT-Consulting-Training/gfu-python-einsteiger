## Datentypen

Jeder **Wert** hat einen Typ. `type()` verrät ihn:

````python
print(type(42))
print(type(3.1415))
print(type("Hallo"))
print(type(True))
print(type(None))
````

| Typ | Bedeutung | Beispiele |
| --- | --- | --- |
| `int` | ganze Zahl, beliebig groß | `42`, `-7`, `1_000_000` |
| `float` | Kommazahl | `3.14`, `-0.5`, `1e6` |
| `str` | Text | `"Hallo"`, `'Welt'` |
| `bool` | Wahrheitswert | `True`, `False` |
| `NoneType` | „kein Wert“ | `None` |

### Vorsicht bei Kommazahlen

`float` speichert Zahlen binär und damit nicht immer exakt:

````python
0.1 + 0.2
````

Für die Ausgabe rundest du einfach, etwa mit `round(x, 2)` oder mit f-Strings (weiter unten). Für Spezialfälle mit exakten Geldbeträgen gibt es das Modul `decimal`; im Kurs brauchen wir es nicht.

## Typumwandlung

Mit `int()`, `float()`, `str()` und `bool()` wandelst du Werte um.

````python
print(int("42") + 1)
print(float("3.5") * 2)
print(str(42) + " ist die Antwort")
print(int(3.99))  # schneidet ab, rundet nicht
print(round(3.99))  # rundet auf eine ganze Zahl: 4
print(round(3.14159, 2))  # rundet auf 2 Nachkommastellen: 3.14
print(round(2.5))  # 2: genau .5 rundet Python zur geraden Zahl
````

Python wandelt **nicht** automatisch zwischen Text und Zahl um. *Explicit is better than implicit:* Du entscheidest, was gemeint ist.

````python
"Alter: " + 18
````

````python
int("drei")
````

## Formatierte Ausgabe mit f-Strings

Ein f-String beginnt mit `f`. In geschweiften Klammern steht beliebiger Python-Code:

````python
name = "Michael"
age = 18
print(f"Hallo {name}, du bist {age} Jahre alt.")
print(f"Nächstes Jahr bist du {age + 1}.")
````

Nach einem Doppelpunkt folgt das **Format**:

````python
price = 1234.5678
print(f"Preis: {price:.2f} €")        # 2 Nachkommastellen
print(f"Preis: {price:,.2f} €")       # mit Tausendertrennzeichen, im US-Format: 1,234.57
print(f"Anteil: {0.256:.1%}")         # Prozent
print(f"|{'links':<10}|{'rechts':>10}|")  # Ausrichtung
print(f"Raum {7:02d}")                # ganze Zahl, 2 Stellen mit führender Null: 07
print(f"Saldo: {-12.5:+.2f} €")       # + zeigt immer ein Vorzeichen: -12.50
print(f"Saldo: {30:+.2f} €")          # +30.00
````

Die Formatangabe `{price:,.2f}` Stück für Stück:

![Die Teile von {price:,.2f}: price ist der Wert, der Doppelpunkt leitet das Format ein, das Komma setzt Tausendertrennzeichen, .2 rundet auf zwei Nachkommastellen, f steht für Kommazahl; Ergebnis Preis: 1,234.57 €.](../assets/01_fstring_format.png)

Zum Debuggen praktisch: `=` zeigt Name und Wert.

````python
print(f"{age=}")
````

> Ältere Schreibweisen wie `"%s" % name` oder `"{}".format(name)` begegnen dir in altem Code. Für neuen Code nimmst du f-Strings.

## Mehrfachzuweisung

Links und rechts vom `=` dürfen mehrere Werte stehen, durch Kommas getrennt. Python verteilt sie der Reihe nach auf die Namen. Diese **Mehrfachzuweisung** heißt auch **Entpacken**; im Kurs sagen wir ab jetzt meist Entpacken.

````python
x, y, z = 1, 2, 3
print(x, y, z)

# Werte tauschen, ganz ohne Hilfsvariable
first_name, last_name = "Huber", "Stefan"
first_name, last_name = last_name, first_name
print(f"Hallo {first_name} {last_name}!")
````

So tauscht `first_name, last_name = last_name, first_name` die Werte:

![Python wertet zuerst die rechte Seite aus und erhält "Stefan" und "Huber", dann verteilt es die Werte der Reihe nach auf first_name und last_name.](../assets/01_werte_tauschen.png)

## Zusammenfassung

- Variablen entstehen durch Zuweisung: `name = wert`.
- Grundtypen: `int`, `float`, `str`, `bool`, `None`.
- Umwandeln mit `int()`, `float()`, `str()`, `bool()`.
- Runden mit `round(x)` bzw. `round(x, n)` auf `n` Nachkommastellen; genau `.5` rundet zur geraden Zahl.
- `float` rechnet nicht immer exakt (`0.1 + 0.2`); für die Ausgabe rundest du.
- Ausgaben mit f-Strings: `f"{wert:.2f}"`.
- Entpacken: `a, b = 1, 2` verteilt mehrere Werte auf mehrere Namen, `a, b = b, a` tauscht sie.
