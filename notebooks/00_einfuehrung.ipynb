## Was ist Python?

Python ist eine **interpretierte**, **dynamisch typisierte** Allzweck-Sprache. Sie wurde 1991 von Guido van Rossum veröffentlicht und ist heute eine der meistgenutzten Sprachen der Welt.

Typische Einsatzgebiete:

- Automatisierung von Routineaufgaben (Dateien, Excel, E-Mails)
- Datenanalyse und Machine Learning
- Web-Backends und APIs
- Skripte für Administration und DevOps

## Python im Vergleich zu Java, C# und C++

| | Python | Java / C# / C++ |
| --- | --- | --- |
| Ausführung | Interpreter, kein separater Compile-Schritt | Compiler (bzw. JIT) |
| Typen | dynamisch: der Wert hat einen Typ, die Variable nicht | statisch: Typ wird bei der Variable deklariert |
| Blöcke | Einrückung (4 Leerzeichen) | geschweifte Klammern `{ }` |
| Semikolon | nicht nötig | Pflicht |
| `main`-Methode | nicht nötig, Code läuft von oben nach unten | Pflicht |

Das klassische „Hello World“ zeigt den Unterschied am deutlichsten. In Java braucht man eine Klasse und eine `main`-Methode, in Python genügt eine Zeile:

````python
print("Hello World!")
````

## Die Python-Art: Zen und Stil

Python hat eine eigene Philosophie, das **Zen of Python**. Es ist sogar in die Sprache eingebaut:

````python
import this
````

Vier Sätze davon begleiten uns durch den Kurs:

- ***Explicit is better than implicit.*** Schreib hin, was du meinst, statt dich auf Magie zu verlassen. Python wandelt z. B. nie heimlich Text in Zahlen um, das machst du ausdrücklich mit `int("42")`.
- *Readability counts.* Code wird viel öfter gelesen als geschrieben.
- *There should be one obvious way to do it.*
- *Errors should never pass silently.* Fehler lieber sichtbar machen als verschlucken.

### Einrückung statt Klammern

Python erkennt Blöcke an der **Einrückung**, nicht an geschweiften Klammern. Daran wird sich auch nichts ändern, wie Python selbst augenzwinkernd klarstellt:

````python
from __future__ import braces
````

Regel: **4 Leerzeichen** pro Ebene, keine Tabs. VS Code macht das automatisch.

### Namenskonventionen (PEP 8)

[PEP 8](https://peps.python.org/pep-0008/) ist der offizielle Stil-Leitfaden. Wer sich daran hält, schreibt Code, den jede Python-Entwicklerin sofort lesen kann.

| Was | Schreibweise | Beispiel |
| --- | --- | --- |
| Variablen, Funktionen | `snake_case` | `net_price`, `calculate_total()` |
| Dateien, Module | `snake_case` | `invoice_tools.py` |
| Klassen | `PascalCase` | `BankAccount` |
| Konstanten | `UPPER_CASE` | `MAX_RETRIES = 3` |
| „intern, nicht anfassen“ | `_führender_unterstrich` | `_cache` |

- **Kein camelCase** wie in Java (`netPrice`). Das funktioniert, wirkt in Python aber fremd.
- Namen auf **Englisch** und sprechend: `customer_count` statt `cc` oder `anzahl_kunden`.
- Konstanten sind nur eine Konvention. Python verhindert nicht, dass man sie ändert.

````python
MAX_PARTICIPANTS = 12  # Konstante
course_name = "Python für Einsteiger"  # Variable
print(f"{course_name}: maximal {MAX_PARTICIPANTS} Teilnehmende")
````

### Stil automatisch prüfen mit ruff

Niemand muss PEP 8 auswendig lernen. Das Werkzeug **ruff** prüft und formatiert den Code automatisch:

```
uv run ruff check .      # Stil- und Fehlerprüfung
uv run ruff format .     # automatisch formatieren
```

In VS Code übernimmt das die Erweiterung *Ruff* beim Speichern.
