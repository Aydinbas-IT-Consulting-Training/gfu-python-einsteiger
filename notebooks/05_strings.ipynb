## Strings anlegen

Einfache und doppelte Anführungszeichen sind gleichwertig. Drei Anführungszeichen erlauben mehrzeiligen Text.

```python
greeting = "Hallo"
quote = 'Sie sagte: "Python ist toll"'
poem = """Erste Zeile
Zweite Zeile"""
print(quote)
print(poem)
```

Manche Zeichen schreibst du mit einem Backslash (**Escape-Sequenz**): `\n` beginnt eine neue Zeile, `\t` setzt einen Tabulator, `\\` steht für den Backslash selbst.

```python
print("Name:\tAda\nOrt:\tKiel")
print("C:\\Users\\ada")
```

`len()` liefert die Länge, `+` verbindet, `*` wiederholt:

```python
print(len(greeting))
print(greeting + " Welt")
print("-" * 20)
```

## Index und Slicing

Jedes Zeichen hat eine Position, beginnend bei **0**. Negative Indizes zählen von hinten.

![Der String "Python" in sechs Kästchen: oben die Indizes 0 bis 5, unten die negativen Indizes -6 bis -1; word[0] ist "P", word[-1] ist "n".](../assets/05_string_index.png)

```python
word = "Python"
print(word[0], word[-1])
```

Mit `[start:stop:step]` schneidest du Teile heraus. `stop` ist wie bei `range()` ausgeschlossen. Am einfachsten denkst du dir die Zahlen als Schnittstellen **zwischen** den Zeichen:

![word[1:4] schneidet an den Schnittstellen 1 und 4 und liefert "yth": Start 1 gehört dazu, Ende 4 nicht mehr, also 4 - 1 = 3 Zeichen.](../assets/05_string_slicing.png)

```python
print(word[0:2])   # "Py"
print(word[2:])    # ab Index 2 bis Ende
print(word[:-2])   # alles außer den letzten zwei
print(word[::-1])  # umgedreht
```

## Wichtige Methoden

```python
text = "  Einstieg in Python  "
print(repr(text.strip()))         # Leerzeichen am Rand entfernen; repr zeigt die Anführungszeichen mit
print(text.upper())
print(text.lower())
print(text.replace("Python", "Java"))
print(text.strip().startswith("Ein"))
print("Python" in text)           # Enthalten-Test
print(text.count("i"))
print(text.find("Python"))        # Position oder -1
```

Prüfmethoden helfen bei Eingaben:

```python
print("2026".isdigit(), "abc".isalpha(), "Abc".istitle())
```

## Zerlegen und Verbinden

`split()` zerlegt einen String in eine Liste, `join()` fügt eine Liste zusammen. `splitlines()` zerlegt mehrzeiligen Text in seine Zeilen.

```python
line = "Ada,Lovelace,1815"
parts = line.split(",")
print(parts)

words = ["Python", "ist", "toll"]
print(" ".join(words))

notes = "Milch kaufen\nBrot kaufen\nZahnarzt anrufen"
print(notes.splitlines())
```
