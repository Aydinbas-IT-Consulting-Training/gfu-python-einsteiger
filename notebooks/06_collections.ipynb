## Listen

````python
members = ["Michael", "Tim", "Werner"]
print(members[0], members[-1], len(members))
````

Listen sind veränderbar:

````python
members.append("Andreas")             # ein Element anhängen
members.extend(["Lili", "Jana"])      # mehrere anhängen
members.insert(0, "Natalija")         # an Position einfügen
members.remove("Tim")                 # nach Wert entfernen
last = members.pop()                  # letztes entfernen und zurückgeben
print(members, last)
````

````python
print("Jana" in members)
print(members[1:3])  # Slicing wie bei Strings
````

### Sortieren

`sorted()` liefert eine **neue** sortierte Liste, `.sort()` sortiert die Liste **selbst**.

````python
numbers = [5, 2, 9, 1]
print(sorted(numbers))
print(sorted(numbers, reverse=True))
print(numbers)  # unverändert

numbers.sort()
print(numbers)  # jetzt sortiert
````

Mit `key=` bestimmst du, **wonach** sortiert wird. Hier nach der Länge der Namen:

````python
names = ["Tim", "Andreas", "Lili", "Michael"]
sorted(names, key=len)
````

Für eigene Kriterien schreibt man oft eine kleine **Lambda-Funktion** direkt an Ort und Stelle. Hier: nach dem letzten Buchstaben.

````python
sorted(names, key=lambda name: name[-1])
````

> `lambda x: ausdruck` ist eine Funktion ohne Namen. Mehr brauchst du davon für den Anfang nicht. Für alles Längere schreibst du eine normale Funktion (Notebook 07).

### Achtung: Zuweisung kopiert nicht

`b = a` legt **keine Kopie** an. Beide Namen zeigen auf **dieselbe** Liste. Änderst du sie über einen Namen, sieht der andere die Änderung auch:

````python
original = [1, 2, 3]
alias = original
alias.append(4)
print(original)
print(id(original) == id(alias), alias is original)
````

`id()` liefert die Identität eines Objekts, `is` prüft, ob zwei Namen auf dasselbe Objekt zeigen. `==` vergleicht dagegen nur den Inhalt.

Bei Zahlen und Strings fällt das nie auf. Sie sind **unveränderlich**: `+=` erzeugt ein neues Objekt und hängt nur den einen Namen um.

````python
a = 10
b = a
b += 1
print(a, b, a is b)
````

**Veränderlich (mutable) oder unveränderlich (immutable)?**

| unveränderlich | veränderlich |
| --- | --- |
| `int`, `float`, `bool`, `str`, `tuple`, `None` | `list`, `dict`, `set`, eigene Objekte |
| „Ändern“ erzeugt ein **neues** Objekt | Methoden ändern **dasselbe** Objekt |

Eine Kuriosität am Rande: CPython hält kleine Zahlen von -5 bis 256 und viele kurze Strings nur **einmal** im Speicher vor und verwendet sie wieder. Weil sie unveränderlich sind, ist das gefahrlos.

````python
print(int("256") is int("256"))  # dasselbe Objekt aus dem Cache
print(int("257") is int("257"))  # zwei verschiedene Objekte
print("hallo" is "hallo")        # kurzer String: wiederverwendet
print("".join(["hal", "lo"]) is "hallo", "".join(["hal", "lo"]) == "hallo")
````

Das ist ein Implementierungsdetail und kann sich ändern. Die Lehre daraus: **Werte immer mit `==` vergleichen.** `is` nur für `x is None`.

Brauchst du eine echte Kopie, sag es ausdrücklich:

````python
original = [1, 2, 3]
copy_1 = original.copy()
copy_2 = list(original)
copy_3 = original[:]
copy_1.append(99)
print(original, copy_1)
print(copy_1 == original[:3], copy_1 is original)
````

Diese Kopien sind **flach**: Enthält die Liste selbst Listen, werden die inneren nicht mitkopiert. Dafür gibt es `copy.deepcopy`:

````python
import copy

matrix = [[1, 2], [3, 4]]
shallow = matrix.copy()
deep = copy.deepcopy(matrix)
matrix[0].append(99)
print(shallow)  # innere Liste geteilt
print(deep)     # unabhängig
````

Dasselbe gilt für Funktionen: Eine übergebene Liste ist **dieselbe** Liste. Ändert die Funktion sie, ändert sie das Original.

````python
def add_bonus(scores):
    scores.append(100)


my_scores = [80, 90]
add_bonus(my_scores)
my_scores
````

## Dictionaries

Ein Dictionary ordnet **Schlüssel** zu **Werten** zu, wie ein Wörterbuch.

````python
capitals = {"CH": "Bern", "DE": "Berlin", "GB": "London"}
print(capitals["DE"])

capitals["FR"] = "Paris"   # hinzufügen oder ändern
del capitals["GB"]         # entfernen
print(capitals)
````

Fehlende Schlüssel führen zu einem `KeyError`. `.get()` liefert stattdessen einen Standardwert:

````python
capitals["US"]
````

````python
capitals.get("US", "unbekannt")
````

Über ein Dictionary iterieren:

````python
for code, city in capitals.items():
    print(f"{code}: {city}")
````

Dictionaries beschreiben oft einen Datensatz; eine Liste davon ist eine kleine Tabelle:

````python
people = [
    {"name": "Michael", "city": "Zürich", "age": 55},
    {"name": "Tim", "city": "Kiel", "age": 34},
]
for person in people:
    print(f"{person['name']} wohnt in {person['city']}")
````

Seit Python 3.15 gibt es auch ein **unveränderliches Dictionary**: `frozendict`. Es passt zu Daten, die sich nicht ändern sollen, etwa Voreinstellungen. Lesen geht wie gewohnt, Ändern führt zu einem `TypeError`:

````python
DEFAULTS = frozendict(language="de", currency="EUR")
print(DEFAULTS["currency"])
DEFAULTS["currency"] = "CHF"
````

## Sets

Ein Set enthält jeden Wert nur einmal.

````python
cities = ["Kiel", "Hamburg", "Zürich", "Kiel", "Hamburg"]
unique_cities = set(cities)
unique_cities
````

````python
python_fans = {"Ada", "Tim", "Lili"}
java_fans = {"Tim", "Werner"}
print(python_fans & java_fans)  # Schnittmenge
print(python_fans | java_fans)  # Vereinigung
print(python_fans - java_fans)  # Differenz
````

## Listen kompakt erzeugen

Eine **List Comprehension** baut eine neue Liste aus einer bestehenden:

````python
numbers = [1, 2, 3, 4, 5, 6]
squares = [n ** 2 for n in numbers]
even = [n for n in numbers if n % 2 == 0]
print(squares, even)
````

Das ist dasselbe wie:

````python
squares = []
for n in numbers:
    squares.append(n ** 2)
squares
````

Comprehensions gibt es auch für Dictionaries (`{schlüssel: wert for …}`) und Sets (`{wert for …}`):

````python
names = ["Ada", "Grace", "Linus"]
name_lengths = {name: len(name) for name in names}
first_letters = {name[0] for name in names}
print(name_lengths)
print(first_letters)
````

Seit Python 3.15 darf vorne in einer Comprehension auch `*` stehen: Es packt jedes Element aus. So machst du aus einer Liste von Listen eine flache Liste:

````python
groups = [["Ada", "Grace"], ["Linus"], ["Guido", "Tim"]]
everyone = [*group for group in groups]
everyone
````

> Halte Comprehensions kurz. Passt eine nicht mehr bequem in eine Zeile oder braucht sie mehrere `if`, ist eine normale Schleife lesbarer.

## Iteratoren: was `for` unter der Haube macht

Listen, Tupel, Strings, Dictionaries, Sets, `range()`, Dateien: Über all das kann `for` laufen. Solche Objekte heißen **iterierbar** (*iterable*). Wie funktioniert das?

`for` holt sich mit `iter()` einen **Iterator**. Das ist ein Objekt, das sich merkt, wo es gerade steht, und mit `next()` das nächste Element liefert:

````python
letters = ["a", "b", "c"]
iterator = iter(letters)
print(next(iterator))
print(next(iterator))
print(next(iterator))
````

Sind alle Elemente geliefert, löst `next()` eine `StopIteration` aus. Genau darauf wartet `for` und beendet die Schleife. In einer `for`-Schleife siehst du diesen Fehler deshalb nie.

````python
next(iterator)
````

Eine `for`-Schleife ist also nichts anderes als:

```
iterator = iter(letters)
while True:
    try:
        letter = next(iterator)
    except StopIteration:
        break
    print(letter)
```

**Wichtig:** Ein Iterator ist nach einem Durchlauf **verbraucht**. Viele eingebaute Funktionen liefern Iteratoren statt Listen, zum Beispiel `zip()` (verbindet Folgen paarweise), `map()`, `filter()` und `reversed()`. Das spart Speicher, kann aber überraschen:

````python
pairs = zip(["Ada", "Grace"], [1815, 1906])
print(list(pairs))
print(list(pairs))  # leer: der Iterator ist schon verbraucht
````

Brauchst du die Werte mehrmals, wandle sie einmal mit `list()` um. Eigene Iteratoren schreibst du am einfachsten als **Generator**, das zeigt Notebook 07.

## Zusammenfassung

- `list`: geordnet, veränderbar: `append`, `extend`, `remove`, `pop`.
- `b = a` kopiert nicht. Kopie mit `a.copy()`, verschachtelt mit `copy.deepcopy(a)`.
- `tuple`: unveränderlich, ideal zum Entpacken: `x, y = point`.
- `dict`: `d[key]`, `d.get(key, default)`, `d.items()`. Unveränderlich: `frozendict`.
- `set`: keine Duplikate, Mengenoperationen `&`, `|`, `-`.
- `sorted(daten, key=..., reverse=True)` sortiert nach eigenem Kriterium.
- Comprehensions für Listen, Dictionaries und Sets; `[*x for x in listen]` macht flach.
- `for` nutzt `iter()` und `next()`. Iteratoren wie `zip()` sind nach einem Durchlauf verbraucht.
