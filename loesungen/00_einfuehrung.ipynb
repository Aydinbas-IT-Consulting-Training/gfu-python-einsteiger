## Was ist Python?

Python ist eine **interpretierte**, **dynamisch typisierte** Allzweck-Sprache. Sie wurde 1991 von Guido van Rossum veröffentlicht und ist heute eine der meistgenutzten Sprachen der Welt.

Typische Einsatzgebiete:

- Automatisierung von Routineaufgaben (Dateien, Excel, E-Mails)
- Datenanalyse und Machine Learning
- Web-Backends und APIs
- Skripte für Administration und DevOps

## Python im Vergleich zu Java, C# und C++

| | Python | Java / C# / C++ |
| --- | --- | --- |
| Ausführung | Interpreter, kein separater Compile-Schritt | Compiler (bzw. JIT) |
| Typen | dynamisch: der Wert hat einen Typ, die Variable nicht | statisch: Typ wird bei der Variable deklariert |
| Blöcke | Einrückung (4 Leerzeichen) | geschweifte Klammern `{ }` |
| Semikolon | nicht nötig | Pflicht |
| `main`-Methode | nicht nötig, Code läuft von oben nach unten | Pflicht |

Das klassische „Hello World“ zeigt den Unterschied am deutlichsten. In Java braucht man eine Klasse und eine `main`-Methode, in Python genügt eine Zeile:

````python
print("Hello World!")
````

## Notebooks: Code in Zellen

Dieses Dokument ist ein **Jupyter-Notebook**. Es besteht aus Zellen:

- **Markdown-Zellen** enthalten Text wie diesen.
- **Code-Zellen** enthalten Python-Code. Ausführen mit `Shift + Enter`.

Der Wert der **letzten Zeile** einer Zelle wird automatisch angezeigt:

````python
7 * 6
````

Mit `print()` gibst du gezielt etwas aus, auch mehrere Werte auf einmal:

````python
result = 7 * 6
print("7 * 6 =", result)
print("Willkommen zum Python-Kurs")
````

> **Achtung:** Zellen teilen sich ihren Zustand. Eine Variable existiert erst, wenn die Zelle mit ihrer Zuweisung ausgeführt wurde. Wenn etwas komisch wirkt: Menü *Kernel → Restart* und alle Zellen von oben neu ausführen.

## Kommentare

Alles hinter `#` ignoriert Python. Kommentare erklären das **Warum**, nicht das Was.

````python
# Mehrwertsteuersatz in Deutschland (Stand 2026)
vat_rate = 0.19
net_price = 100
print(net_price * (1 + vat_rate))  # Bruttopreis
````

## Die eingebaute Hilfe

`help()` zeigt die Dokumentation zu einer Funktion. In Notebooks geht es auch kürzer mit einem `?` hinter dem Namen.

````python
help(print)
````

````python
print?
````

## Module: fertige Bausteine nutzen

Python bringt eine große **Standardbibliothek** mit. Mit `import` holst du dir ein Modul, zum Beispiel für Zufallszahlen:

````python
import random

print(random.randint(1, 6))  # Würfelwurf
````

## Zusammenfassung

- Python wird interpretiert, Blöcke entstehen durch Einrückung.
