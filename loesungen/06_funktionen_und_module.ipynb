## Warum Funktionen?

Funktionen geben einem Stück Code einen **Namen**. Das macht Programme kürzer, lesbarer und testbar. Faustregel: Kopierst du Code ein zweites Mal, gehört er in eine Funktion.

## Funktionen definieren

````python
def greet(name):
    print(f"Hallo {name}!")


greet("Ada")
greet("Linus")
````

Mit `return` liefert eine Funktion ein Ergebnis zurück. Ohne `return` liefert sie `None`.

````python
def gross_price(net_price, vat_rate):
    return net_price * (1 + vat_rate)


price = gross_price(100, 0.19)
print(f"{price:.2f} €")
````

> `print` zeigt etwas an, `return` gibt etwas zurück. Nur mit `return` kannst du mit dem Ergebnis weiterrechnen.

## Standardwerte und Schlüsselwort-Argumente

Parameter können einen **Standardwert** haben. Beim Aufruf darfst du Argumente auch **mit Namen** übergeben; dann ist die Reihenfolge egal.

````python
def gross_price(net_price, vat_rate=0.19):
    return net_price * (1 + vat_rate)


print(gross_price(100))
print(gross_price(100, 0.07))
print(gross_price(vat_rate=0.07, net_price=100))
````

> **Typischer Fehler:** Kein veränderbarer Standardwert wie `def f(items=[])`. Die Liste wird zwischen Aufrufen geteilt. Nimm stattdessen `None` und lege die Liste in der Funktion an.

## Mehrere Rückgabewerte

Eine Funktion kann ein Tupel zurückgeben, das du direkt entpackst:

````python
def min_max(numbers):
    return min(numbers), max(numbers)


lowest, highest = min_max([5, 2, 9, 1])
print(lowest, highest)
````

## Docstrings und Type Hints

Ein **Docstring** beschreibt, was die Funktion tut. **Type Hints** zeigen, welche Typen erwartet werden. Python prüft sie nicht, aber Editoren und Kollegen profitieren davon.

````python
def bmi(weight_kg: float, height_cm: float) -> float:
    """Berechnet den Body-Mass-Index aus Gewicht und Größe."""
    return weight_kg / (height_cm / 100) ** 2


help(bmi)
````

## Module

Jede `.py`-Datei ist ein **Modul**. Aus der Standardbibliothek kennst du schon `random`. Es gibt mehrere Import-Varianten:

````python
import math
from random import randint
from datetime import date as Date

print(math.sqrt(16))
print(randint(1, 6))
print(Date.today())
````

## Übungen

**Übung 1 (Würfelspiel):** Schreibe `roll_dice()`, die eine Zahl von 1 bis 6 liefert, und `play_until_six()`, die so lange würfelt, bis eine 6 fällt, und die Anzahl der Würfe **zurückgibt**.

````python
import random


def roll_dice() -> int:
    return random.randint(1, 6)


def play_until_six() -> int:
    tries = 0
    while True:
        tries += 1
        if roll_dice() == 6:
            return tries


print(f"Eine 6 nach {play_until_six()} Würfen")
````

**Übung 2 (Taschenrechner):** Schreibe `calculate(x, y, operator)` für `+`, `-`, `*` und `/`. Bei einem unbekannten Operator soll die Funktion `None` zurückgeben.

````python
def calculate(x: float, y: float, operator: str) -> float | None:
    if operator == "+":
        return x + y
    if operator == "-":
        return x - y
    if operator == "*":
        return x * y
    if operator == "/":
        return x / y
    return None


print(calculate(6, 7, "*"))
print(calculate(6, 7, "^"))
````

**Übung 3 (Temperatur):** Schreibe `to_celsius(fahrenheit)` und `to_fahrenheit(celsius)` mit Type Hints und Docstring. Prüfe, dass `to_celsius(to_fahrenheit(20))` wieder `20` ergibt.

````python
def to_celsius(fahrenheit: float) -> float:
    """Rechnet Fahrenheit in Celsius um."""
    return (fahrenheit - 32) * 5 / 9


def to_fahrenheit(celsius: float) -> float:
    """Rechnet Celsius in Fahrenheit um."""
    return celsius * 9 / 5 + 32


print(to_celsius(to_fahrenheit(20)))
````

**Übung 4 (Refactoring):** Der folgende Code wiederholt sich. Pack die Logik in eine Funktion `format_invoice_line(product, quantity, unit_price)` und rufe sie dreimal auf.

````python
print(f"{'Kaffee':<10} {2:>3} x {4.50:>6.2f} € = {2 * 4.50:>7.2f} €")
print(f"{'Kuchen':<10} {1:>3} x {3.80:>6.2f} € = {1 * 3.80:>7.2f} €")
print(f"{'Wasser':<10} {3:>3} x {2.20:>6.2f} € = {3 * 2.20:>7.2f} €")
````

````python
def format_invoice_line(product: str, quantity: int, unit_price: float) -> str:
    total = quantity * unit_price
    return f"{product:<10} {quantity:>3} x {unit_price:>6.2f} € = {total:>7.2f} €"


print(format_invoice_line("Kaffee", 2, 4.50))
print(format_invoice_line("Kuchen", 1, 3.80))
print(format_invoice_line("Wasser", 3, 2.20))
````

## Zusammenfassung

- `import modul`, `from modul import name`; Startcode in `if __name__ == "__main__":`.
