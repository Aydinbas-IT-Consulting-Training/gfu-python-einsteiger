Für jeden Schritt gibt es mehrere Wege. Probiere jeweils mindestens zwei aus und entscheide, welchen du nimmst, und warum:

- Mehrere ergänzen: `extend`, `+=` oder `shopping + [...]`. Und was macht `append` mit einer Liste?
- Vorne einfügen: `insert(0, ...)` oder `["Kaffee"] + shopping`.
- Abhaken: `remove` (nach Wert), `del shopping[i]` (nach Position) oder eine Comprehension, die alle Vorkommen herausfiltert.

Untersuchen: Welche Ausgabe erzeugt `shopping.append(["Eier", "Mehl"])`, wie lang ist die Liste danach, und warum? Welche Wege ändern die bestehende Liste, welche erzeugen eine neue? Was passiert bei `shopping.remove("Butter")`, wenn Butter nicht auf der Liste steht (Zelle unten)?

Ändern: Entferne einen Eintrag nur, wenn er vorhanden ist.

````python
shopping = ["Milch", "Brot"]
````

````python
shopping.remove("Butter")
````

````python
new_items = ["Eier", "Mehl"]

# Mehrere ergänzen: drei Wege mit demselben Ergebnis
a = ["Milch", "Brot"]
a.extend(new_items)  # ändert die bestehende Liste
b = ["Milch", "Brot"]
b += new_items  # ändert die bestehende Liste, wie extend
c = ["Milch", "Brot"] + new_items  # erzeugt eine neue Liste
print(a, b, c)

# Falle: append hängt die ganze Liste als EIN Element an
d = ["Milch", "Brot"]
d.append(new_items)
print(d, len(d))  # ['Milch', 'Brot', ['Eier', 'Mehl']] 3

# Vorne einfügen: zwei Wege
e = ["Milch", "Brot"]
e.insert(0, "Kaffee")  # ändert die Liste
f = ["Kaffee"] + ["Milch", "Brot"]  # neue Liste
print(e, f)

# Abhaken: drei Wege
g = ["Kaffee", "Milch", "Brot", "Brot"]
g.remove("Brot")  # nach Wert, nur das erste Vorkommen
h = ["Kaffee", "Milch", "Brot", "Brot"]
del h[2]  # nach Position
i = [item for item in ["Kaffee", "Milch", "Brot", "Brot"] if item != "Brot"]
print(g, h, i)  # die Comprehension entfernt alle Vorkommen und baut eine neue Liste

# Die ganze Aufgabe mit den üblichen Wegen
shopping = ["Milch", "Brot"]
shopping.extend(["Eier", "Mehl"])
shopping.insert(0, "Kaffee")
shopping.remove("Brot")
last = shopping.pop()  # den zuletzt ergänzten wieder herausnehmen
if "Milch" not in shopping:  # nur ergänzen, wenn noch nicht drauf
    shopping.append("Milch")
print(shopping, last)

# Wahl: extend, insert und remove sagen am klarsten, was passiert, und ändern die
# Liste selbst. + und die Comprehension bauen eine neue Liste; das ist richtig,
# wenn das Original unverändert bleiben soll.
# remove("Butter") wirft einen ValueError, wenn Butter nicht auf der Liste steht.

# Ändern: nur entfernen, wenn der Eintrag vorhanden ist
if "Butter" in shopping:
    shopping.remove("Butter")
print(shopping)
````

### Sortieren

`sorted()` liefert eine **neue** sortierte Liste, `.sort()` sortiert die Liste **selbst**.

````python
numbers = [5, 2, 9, 1]
print(sorted(numbers))
print(sorted(numbers, reverse=True))
print(numbers)  # unverändert

numbers.sort()
print(numbers)  # jetzt sortiert
````

Mit `key=` bestimmst du, **wonach** sortiert wird. Hier nach der Länge der Namen:

````python
names = ["Tim", "Andreas", "Lili", "Michael"]
sorted(names, key=len)
````

Für eigene Kriterien schreibt man oft eine kleine **Lambda-Funktion** direkt an Ort und Stelle. Hier: nach dem letzten Buchstaben.

````python
sorted(names, key=lambda name: name[-1])
````

> `lambda x: ausdruck` ist eine Funktion ohne Namen. Mehr brauchst du davon für den Anfang nicht. Für alles Längere schreibst du eine normale Funktion (Notebook 07).

### Achtung: Zuweisung kopiert nicht

`b = a` legt **keine Kopie** an. Beide Namen zeigen auf **dieselbe** Liste. Änderst du sie über einen Namen, sieht der andere die Änderung auch:

````python
original = [1, 2, 3]
alias = original
alias.append(4)
print(original)
print(id(original) == id(alias), alias is original)
````

`id()` liefert die Identität eines Objekts, `is` prüft, ob zwei Namen auf dasselbe Objekt zeigen. `==` vergleicht dagegen nur den Inhalt.

Bei Zahlen und Strings fällt das nie auf. Sie sind **unveränderlich**: `+=` erzeugt ein neues Objekt und hängt nur den einen Namen um.

````python
a = 10
b = a
b += 1
print(a, b, a is b)
````

**Veränderlich (mutable) oder unveränderlich (immutable)?**

| unveränderlich | veränderlich |
| --- | --- |
| `int`, `float`, `bool`, `str`, `tuple`, `None` | `list`, `dict`, `set`, eigene Objekte |
| „Ändern“ erzeugt ein **neues** Objekt | Methoden ändern **dasselbe** Objekt |

Eine Kuriosität am Rande: CPython hält kleine Zahlen von -5 bis 256 und viele kurze Strings nur **einmal** im Speicher vor und verwendet sie wieder. Weil sie unveränderlich sind, ist das gefahrlos.

````python
print(int("256") is int("256"))  # dasselbe Objekt aus dem Cache
print(int("257") is int("257"))  # zwei verschiedene Objekte
print("hallo" is "hallo")        # kurzer String: wiederverwendet
print("".join(["hal", "lo"]) is "hallo", "".join(["hal", "lo"]) == "hallo")
````

Das ist ein Implementierungsdetail und kann sich ändern. Die Lehre daraus: **Werte immer mit `==` vergleichen.** `is` nur für `x is None` (warum, steht in Notebook 03, Abschnitt „None und Vergleiche: `==` oder `is`?“).

Brauchst du eine echte Kopie, sag es ausdrücklich:

````python
original = [1, 2, 3]
copy_1 = original.copy()
copy_2 = list(original)
copy_3 = original[:]
copy_1.append(99)
print(original, copy_1)
print(copy_1 == original[:3], copy_1 is original)
````

Diese Kopien sind **flach**: Enthält die Liste selbst Listen, werden die inneren nicht mitkopiert. Dafür gibt es `copy.deepcopy`:

````python
import copy

matrix = [[1, 2], [3, 4]]
shallow = matrix.copy()
deep = copy.deepcopy(matrix)
matrix[0].append(99)
print(shallow)  # innere Liste geteilt
print(deep)     # unabhängig
````

Dasselbe gilt für Funktionen: Eine übergebene Liste ist **dieselbe** Liste. Ändert die Funktion sie, ändert sie das Original.

````python
def add_bonus(scores):
    scores.append(100)


my_scores = [80, 90]
add_bonus(my_scores)
my_scores
````

## Dictionaries

Ein Dictionary ordnet **Schlüssel** zu **Werten** zu, wie ein Wörterbuch.

````python
capitals = {"CH": "Bern", "DE": "Berlin", "GB": "London"}
print(capitals["DE"])

capitals["FR"] = "Paris"   # hinzufügen oder ändern
del capitals["GB"]         # entfernen
print(capitals)
````

Fehlende Schlüssel führen zu einem `KeyError`. `.get()` liefert stattdessen einen Standardwert:

````python
capitals["US"]
````

````python
capitals.get("US", "unbekannt")
````

Über ein Dictionary iterieren:

````python
for code, city in capitals.items():
    print(f"{code}: {city}")
````

Dictionaries beschreiben oft einen Datensatz; eine Liste davon ist eine kleine Tabelle:

````python
people = [
    {"name": "Michael", "city": "Zürich", "age": 55},
    {"name": "Tim", "city": "Kiel", "age": 34},
]
for person in people:
    print(f"{person['name']} wohnt in {person['city']}")
````

Seit Python 3.15 gibt es auch ein **unveränderliches Dictionary**: `frozendict`. Es passt zu Daten, die sich nicht ändern sollen, etwa Voreinstellungen. Lesen geht wie gewohnt, Ändern führt zu einem `TypeError`:

````python
DEFAULTS = frozendict(language="de", currency="EUR")
print(DEFAULTS["currency"])
DEFAULTS["currency"] = "CHF"
````

## Listen kompakt erzeugen

Eine **List Comprehension** baut eine neue Liste aus einer bestehenden:

````python
numbers = [1, 2, 3, 4, 5, 6]
squares = [n ** 2 for n in numbers]
even = [n for n in numbers if n % 2 == 0]
print(squares, even)
````

Das ist dasselbe wie:

````python
squares = []
for n in numbers:
    squares.append(n ** 2)
squares
````

Comprehensions gibt es auch für Dictionaries (`{schlüssel: wert for …}`) und Sets (`{wert for …}`):

````python
names = ["Ada", "Grace", "Linus"]
name_lengths = {name: len(name) for name in names}
first_letters = {name[0] for name in names}
print(name_lengths)
print(first_letters)
````

Seit Python 3.15 darf vorne in einer Comprehension auch `*` stehen: Es packt jedes Element aus. So machst du aus einer Liste von Listen eine flache Liste:

````python
groups = [["Ada", "Grace"], ["Linus"], ["Guido", "Tim"]]
everyone = [*group for group in groups]
everyone
````

> Halte Comprehensions kurz. Passt eine nicht mehr bequem in eine Zeile oder braucht sie mehrere `if`, ist eine normale Schleife lesbarer.

**Übung 3 (Notenrechner, leicht):** Berechne aus `grades` den Durchschnitt, die beste und die schlechteste Note sowie die Zahl der bestandenen Prüfungen (bis 4,0).

- Schreibe das zuerst selbst mit einer `for`-Schleife, ohne `sum`, `min`, `max` und `len`.
- Nutze dann die eingebauten Funktionen `sum`, `min`, `max` und `len` sowie für die bestandenen Prüfungen eine Comprehension, und vergleiche beide Fassungen: Liefern sie dasselbe? Welche liest sich besser?
- Gib außerdem die drei besten Noten aus, mit `sorted` und Slicing.

Was passiert bei einer leeren Liste?

````python
grades = [2.3, 1.7, 4.0, 3.3, 5.0, 1.0]
````

````python
# Fassung 1: selbst geschrieben, nur mit einer Schleife
total = 0
count = 0
passed_count = 0
best = grades[0]
worst = grades[0]
for grade in grades:
    total += grade
    count += 1
    if grade < best:
        best = grade
    if grade > worst:
        worst = grade
    if grade <= 4.0:
        passed_count += 1
print(f"Schnitt {total / count:.2f}, beste {best}, schlechteste {worst}")
print(f"{passed_count} von {count} bestanden")

# Fassung 2: mit eingebauten Funktionen
passed = [grade for grade in grades if grade <= 4.0]
average = sum(grades) / len(grades)
print(f"Schnitt {average:.2f}, beste {min(grades)}, schlechteste {max(grades)}")
print(f"{len(passed)} von {len(grades)} bestanden")
print(total == sum(grades), best == min(grades), worst == max(grades))  # True True True

print("Die drei besten:", sorted(grades)[:3])

# Vergleich: Fassung 1 zeigt, was sum, min und max intern tun; Fassung 2 ist kürzer
# und sagt direkt, was gemeint ist. Im Alltag nimmt man Fassung 2.
# Bei Schulnoten ist min() die beste Note.
# Leere Liste: sum([]) / len([]) wirft ZeroDivisionError, min([]) einen ValueError;
# Fassung 1 scheitert schon an grades[0] mit einem IndexError.
````

## Iteratoren: was `for` unter der Haube macht

Listen, Tupel, Strings, Dictionaries, Sets, `range()`, Dateien: Über all das kann `for` laufen. Solche Objekte heißen **iterierbar** (*iterable*). Wie funktioniert das?

`for` holt sich mit `iter()` einen **Iterator**. Das ist ein Objekt, das sich merkt, wo es gerade steht, und mit `next()` das nächste Element liefert:

````python
letters = ["a", "b", "c"]
iterator = iter(letters)
print(next(iterator))
print(next(iterator))
print(next(iterator))
````

Sind alle Elemente geliefert, löst `next()` eine `StopIteration` aus. Genau darauf wartet `for` und beendet die Schleife. In einer `for`-Schleife siehst du diesen Fehler deshalb nie.

````python
next(iterator)
````

Eine `for`-Schleife ist also nichts anderes als:

```
iterator = iter(letters)
while True:
    try:
        letter = next(iterator)
    except StopIteration:
        break
    print(letter)
```

**Wichtig:** Ein Iterator ist nach einem Durchlauf **verbraucht**. Viele eingebaute Funktionen liefern Iteratoren statt Listen, zum Beispiel `zip()` (verbindet Folgen paarweise), `map()`, `filter()` und `reversed()`. Das spart Speicher, kann aber überraschen:

````python
pairs = zip(["Ada", "Grace"], [1815, 1906])
print(list(pairs))
print(list(pairs))  # leer: der Iterator ist schon verbraucht
````

Brauchst du die Werte mehrmals, wandle sie einmal mit `list()` um. Eigene Iteratoren schreibst du am einfachsten als **Generator**, das zeigt Notebook 07.

**Übung 4 (Top-3-Umsätze: zwei Lösungen, mittel):** In `sales` steht der Tagesumsatz je Produkt. Der Startcode enthält zwei korrekte Lösungen für die drei umsatzstärksten Produkte, wie sie auch eine KI vorschlagen könnte. Sage die Ausgabe vorher, führe dann beide aus und entscheide: Welche nimmst du, und warum? Kriterien: Lesbarkeit, Zeilenzahl, was beim Gleichstand von Kuchen und Croissant passiert.

- Ändern: Gib die Top 3 nummeriert mit `enumerate(..., start=1)` und in Euro formatiert aus, etwa `1. Kaffee: 412.50 €`.

````python
sales = [
    ("Kaffee", 412.50),
    ("Tee", 98.00),
    ("Kuchen", 265.40),
    ("Saft", 143.10),
    ("Croissant", 265.40),
]

# Lösung A: dreimal den größten Umsatz suchen und aus einer Kopie entfernen
remaining = sales.copy()
top_a = []
for _ in range(3):
    best = remaining[0]
    for entry in remaining:
        if entry[1] > best[1]:
            best = entry
    top_a.append(best)
    remaining.remove(best)
print(top_a)

# Lösung B
top_b = sorted(sales, key=lambda entry: entry[1], reverse=True)[:3]
print(top_b)
````

````python
for place, (product, amount) in enumerate(top_b, start=1):
    print(f"{place}. {product}: {amount:.2f} €")

# Gleichstand: Beide Lösungen behalten Kuchen vor Croissant (Reihenfolge in sales).
# A nimmt bei > den ersten gefundenen Größten, sorted() ist stabil.
# Mit >= statt > stünde in Lösung A Croissant vor Kuchen.
# Wahl: B ist kürzer und sagt, was es tut; A zeigt, was sorted() dir abnimmt.
````

**Übung 6 (Logzeilen auswerten, schwer):** Der Startcode zerlegt den mehrzeiligen String `log` mit `splitlines()` in eine Liste von Zeilen. Werte die Liste aus und gib das Ergebnis als kleinen Bericht aus:

- Sammle alle `ERROR`-Zeilen per Comprehension, jeweils mit ihrer Zeilennummer (`enumerate(..., start=1)`).
- Gib die Uhrzeit des ersten und des letzten Fehlers aus (über `[0]` und `[-1]`).
- Sammle die Level aller Zeilen in einer Liste `levels` und zähle sie mit `levels.count(level)`.

Untersuchen: Was liefert `lines[1].split()`, und an welcher Position steht das Level?

Ändern: Zähle die Level mit `collections.Counter` und vergleiche mit deiner Zählung. Hinweis: `from collections import Counter`, dann `Counter(levels)` und `.most_common()`.

````python
log = """2026-11-02 08:00:01 INFO Server gestartet
2026-11-02 08:15:02 ERROR Datenbank nicht erreichbar
2026-11-02 08:15:30 INFO Neuer Versuch
2026-11-02 08:16:00 WARNING Antwortzeit 2,3 s
2026-11-02 09:02:11 INFO Anmeldung ada
2026-11-02 09:40:45 ERROR Zeitüberschreitung bei Zahlung
2026-11-02 10:05:00 INFO Bericht erstellt
2026-11-02 11:20:13 WARNING Speicher zu 85 % belegt
2026-11-02 12:01:59 ERROR Datei bericht.pdf fehlt"""

lines = log.splitlines()
print(len(lines), "Zeilen")
````

````python
from collections import Counter

print(lines[1].split())  # Datum, Uhrzeit, Level, Wörter der Meldung: Level an Index 2

errors = [
    (number, line)
    for number, line in enumerate(lines, start=1)
    if line.split()[2] == "ERROR"
]
first_time = errors[0][1].split()[1]
last_time = errors[-1][1].split()[1]
levels = [line.split()[2] for line in lines]

print(f"Bericht: {len(lines)} Zeilen, davon {len(errors)} Fehler")
for number, line in errors:
    print(f"  Zeile {number}: {line}")
print(f"Erster Fehler um {first_time}, letzter um {last_time}")
for level in ["INFO", "WARNING", "ERROR"]:
    print(f"{level}: {levels.count(level)}")

# Ändern: Counter zählt alle Level in einem Schritt, häufigstes zuerst
print(Counter(levels).most_common())
````

## Zusammenfassung

- `list`: geordnet, veränderbar: `append`, `extend`, `insert`, `remove`, `pop`, `count`; `+` baut eine neue Liste.
- `.sort()` ändert die Liste und liefert `None`, `sorted()` liefert eine neue Liste; `key=str.lower` sortiert ohne Rücksicht auf Groß- und Kleinschreibung.
- `b = a` kopiert nicht. Kopie mit `a.copy()`, verschachtelt mit `copy.deepcopy(a)`.
- `tuple`: unveränderlich, ideal zum Entpacken: `x, y = point`.
- `dict`: `d[key]`, `d.get(key, default)`, `d.items()`. Unveränderlich: `frozendict`.
- `set`: keine Duplikate, Mengenoperationen `&`, `|`, `-`.
- `sorted(daten, key=..., reverse=True)` sortiert nach eigenem Kriterium.
- Comprehensions für Listen, Dictionaries und Sets; `[*x for x in listen]` macht flach.
- `collections.Counter` zählt gleiche Einträge, `.most_common()` sortiert nach Häufigkeit.
- `for` nutzt `iter()` und `next()`. Iteratoren wie `zip()` sind nach einem Durchlauf verbraucht.
