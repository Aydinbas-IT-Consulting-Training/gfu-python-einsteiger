### Sortieren

`sorted()` liefert eine **neue** sortierte Liste, `.sort()` sortiert die Liste **selbst**.

```python
numbers = [5, 2, 9, 1]
print(sorted(numbers))
print(sorted(numbers, reverse=True))
print(numbers)  # unverändert

numbers.sort()
print(numbers)  # jetzt sortiert
```

Mit `key=` bestimmst du, **wonach** sortiert wird: `sorted` ruft die angegebene Funktion für jedes Element auf und sortiert nach deren Ergebnis. Du übergibst die Funktion ohne Klammern, also `key=len`, nicht `key=len()`. Hier nach der Länge der Namen:

```python
names = ["Tim", "Andreas", "Lili", "Michael"]
sorted(names, key=len)
```

### Achtung: Zuweisung kopiert nicht

`b = a` legt **keine Kopie** an. Beide Namen zeigen auf **dieselbe** Liste. Änderst du sie über einen Namen, sieht der andere die Änderung auch:

```python
original = [1, 2, 3]
alias = original
alias.append(4)
print(original)
print(id(original) == id(alias), alias is original)
```

`id()` liefert die Identität eines Objekts, `is` prüft, ob zwei Namen auf dasselbe Objekt zeigen. `==` vergleicht dagegen nur den Inhalt.

Bei Zahlen und Strings fällt das nie auf. Sie sind **unveränderlich**: `+=` erzeugt ein neues Objekt und hängt nur den einen Namen um.

```python
a = 10
b = a
b += 1
print(a, b, a is b)
```

**Veränderlich (mutable) oder unveränderlich (immutable)?**

| unveränderlich | veränderlich |
| --- | --- |
| `int`, `float`, `bool`, `str`, `tuple`, `None` | `list`, `dict`, `set`, eigene Objekte |
| „Ändern“ erzeugt ein **neues** Objekt | Methoden ändern **dasselbe** Objekt |

Die Regel für Vergleiche: **Werte immer mit `==` vergleichen.** `is` nur für `x is None` (warum, steht in Notebook 03, Abschnitt „None und Vergleiche: `==` oder `is`?“).

Brauchst du eine echte Kopie, sag es ausdrücklich:

```python
original = [1, 2, 3]
copy_1 = original.copy()
copy_2 = list(original)
copy_3 = original[:]
copy_1.append(99)
print(original, copy_1)
print(copy_1 == original[:3], copy_1 is original)
```

## Listen mit `match` zerlegen

`match` aus dem Exkurs in Notebook 03 vergleicht nicht nur einzelne Werte, sondern auch Listen mit Mustern. Typisches Beispiel: ein Befehl mit Argumenten, wie ihn ein Spiel oder ein Kommandozeilen-Tool einliest. `split()` aus Notebook 05 zerlegt den Text in eine Liste von Wörtern: `"move 3 4".split()` liefert `["move", "3", "4"]`.

`match` vergleicht diese Liste mit jedem Muster: Feste Werte wie `"move"` müssen genau passen, ein Name wie `x` nimmt das Wort an dieser Stelle auf. Probier verschiedene Eingaben aus:

```python
user_input = "move 3 4"  # auch "quit", "move 3" und "jump" probieren

match user_input.split():
    case ["quit"]:
        print("Tschüss")
    case ["move", x, y]:
        print(f"Gehe nach ({x}, {y})")
    case _:
        print("Unbekannter Befehl")
```

So vergleicht `match` die Eingabe `"move 3 4"` Stelle für Stelle mit dem Muster `["move", x, y]`:

![Die Wörter "move", "3" und "4" stehen über dem Muster "move", x, y: "move" muss gleich sein, x bekommt "3", y bekommt "4", das Muster passt und die Ausgabe ist Gehe nach (3, 4).](../assets/06_match_liste.png)

`["move", x, y]` passt nur auf genau drei Wörter, deshalb landet `"move 3"` im Zweig `case _`.

Willst du im Zweig wissen, welcher der Werte hinter `|` gepasst hat, gib ihm mit `as` einen Namen. Das geht bei einem einzelnen Wert wie im Beispiel unten und genauso an einer Stelle in einem Muster für Listen:

```python
command = "halt"

match command:
    case "stop" | "halt" as word:
        print(f"Motor aus (Befehl: {word})")
    case _:
        print("Motor läuft weiter")
```

**Kurzübung 3 (Einkaufsliste per Befehl):** Eine Einkaufslisten-App liest Befehle als Text: `"neu Brot"` setzt Brot auf die Liste `shopping`, `"weg Milch"` streicht Milch, `"zeigen"` gibt die Liste aus. Jeder andere Befehl ergibt `Unbekannter Befehl: ` mit dem Befehl dahinter. Werte die Befehle in `commands` nacheinander in einer `for`-Schleife aus: Zerlege jeden mit `split()` und unterscheide die Fälle mit `match` und Mustern für Listen. Erwartet: `zeigen` gibt `['Brot', 'Eier']` aus, danach erscheint `Unbekannter Befehl: neu`.

- **Untersuchen:** Warum landet `"neu"` ohne Ware im Zweig `case _`?

```python
shopping = ["Milch"]
commands = ["neu Brot", "neu Eier", "weg Milch", "zeigen", "neu"]
```

```python
shopping = ["Milch"]  # bei jedem Lauf neu beginnen
for command in commands:
    match command.split():
        case ["neu", item]:
            shopping.append(item)
        case ["weg", item]:
            shopping.remove(item)
        case ["zeigen"]:
            print(shopping)
        case _:
            print(f"Unbekannter Befehl: {command}")
# zeigen -> ['Brot', 'Eier'], danach: Unbekannter Befehl: neu

# Untersuchen: "neu".split() liefert ["neu"], ein Wort. ["neu", item] passt nur
# auf genau zwei Wörter, also bleibt nur case _.
```

## Dictionaries

Ein Dictionary ordnet **Schlüssel** zu **Werten** zu, wie ein Wörterbuch.

![Das Dictionary capitals als zwei Spalten: die Schlüssel "CH", "DE" und "GB" zeigen auf die Werte "Bern", "Berlin" und "London"; die Abfrage mit dem Schlüssel "DE" liefert "Berlin".](../assets/06_dictionary.png)

```python
capitals = {"CH": "Bern", "DE": "Berlin", "GB": "London"}
print(capitals["DE"])

capitals["FR"] = "Paris"   # hinzufügen oder ändern
del capitals["GB"]         # entfernen
print(capitals)
```

Fehlende Schlüssel führen zu einem `KeyError`. `.get()` liefert stattdessen einen Standardwert:

```python
capitals["US"]
```

```python
capitals.get("US", "unbekannt")
```

`in` prüft bei einem Dictionary, ob ein **Schlüssel** vorhanden ist, nicht ein Wert. `.keys()` liefert alle Schlüssel, `.values()` alle Werte:

```python
print("DE" in capitals)      # True: DE ist ein Schlüssel
print("Berlin" in capitals)  # False: Berlin ist nur ein Wert
print(list(capitals.keys()))
print(list(capitals.values()))
```

Über ein Dictionary iterieren:

```python
for code, city in capitals.items():
    print(f"{code}: {city}")
```

Dictionaries beschreiben oft einen Datensatz; eine Liste davon ist eine kleine Tabelle:

```python
people = [
    {"name": "Michael", "city": "Zürich", "age": 55},
    {"name": "Tim", "city": "Kiel", "age": 34},
]
for person in people:
    print(f"{person['name']} wohnt in {person['city']}")
```

Seit Python 3.15 gibt es auch ein **unveränderliches Dictionary**: `frozendict`. Es passt zu Daten, die sich nicht ändern sollen, etwa Voreinstellungen. Lesen geht wie gewohnt, Ändern führt zu einem `TypeError`:

```python
DEFAULTS = frozendict(language="de", currency="EUR")
print(DEFAULTS["currency"])
DEFAULTS["currency"] = "CHF"
```

## Sets

Ein Set enthält jeden Wert nur einmal. Achtung: `{}` ist ein leeres **Dictionary**; ein leeres Set legst du mit `set()` an.

```python
cities = ["Kiel", "Hamburg", "Zürich", "Kiel", "Hamburg"]
unique_cities = set(cities)
unique_cities
```

```python
python_fans = {"Ada", "Tim", "Lili"}
java_fans = {"Tim", "Werner"}
print(python_fans & java_fans)  # Schnittmenge
print(python_fans | java_fans)  # Vereinigung
print(python_fans - java_fans)  # Differenz
```

## Listen kompakt erzeugen

Eine **List Comprehension** baut eine neue Liste aus einer bestehenden:

```python
numbers = [1, 2, 3, 4, 5, 6]
squares = [n ** 2 for n in numbers]
even = [n for n in numbers if n % 2 == 0]
print(squares, even)
```

Das ist dasselbe wie:

```python
squares = []
for n in numbers:
    squares.append(n ** 2)
squares
```

Comprehensions gibt es auch für Dictionaries (`{schlüssel: wert for …}`) und Sets (`{wert for …}`):

```python
names = ["Ada", "Grace", "Linus"]
name_lengths = {name: len(name) for name in names}
first_letters = {name[0] for name in names}
print(name_lengths)
print(first_letters)
```

Seit Python 3.15 darf vorne in einer Comprehension auch `*` stehen: Es packt jedes Element aus. So machst du aus einer Liste von Listen eine flache Liste:

```python
groups = [["Ada", "Grace"], ["Linus"], ["Guido", "Tim"]]
everyone = [*group for group in groups]
everyone
```

> Halte Comprehensions kurz. Passt eine nicht mehr bequem in eine Zeile oder braucht sie mehrere `if`, ist eine normale Schleife lesbarer.

**Übung 4 (Top-3-Umsätze: zwei Lösungen, mittel):** In `sales` steht der Tagesumsatz je Produkt. Der Startcode enthält zwei korrekte Lösungen für die drei umsatzstärksten Produkte, wie sie auch eine KI vorschlagen könnte. Sage die Ausgabe vorher, führe dann beide aus und entscheide: Welche nimmst du, und warum? Kriterien: Lesbarkeit, Zeilenzahl, was beim Gleichstand von Kuchen und Croissant passiert.

- **Ändern:** Gib die Top 3 nummeriert mit `enumerate(..., start=1)` und in Euro formatiert aus, etwa `1. Kaffee: 412.50 €`.

```python
sales = [
    ("Kaffee", 412.50),
    ("Tee", 98.00),
    ("Kuchen", 265.40),
    ("Saft", 143.10),
    ("Croissant", 265.40),
]

# Lösung A: dreimal den größten Umsatz suchen und aus einer Kopie entfernen
remaining = sales.copy()
top_a = []
for _ in range(3):  # _ : Name für einen Wert, den wir nicht brauchen
    best = remaining[0]
    for entry in remaining:
        if entry[1] > best[1]:
            best = entry
    top_a.append(best)
    remaining.remove(best)
print(top_a)

# Lösung B
# lambda entry: entry[1] = sortiere nach dem Umsatz; mehr in Notebook 07
top_b = sorted(sales, key=lambda entry: entry[1], reverse=True)[:3]
print(top_b)
```

Tipp: Guards (`case ... if ...`) zeigt der Exkurs zu `match` in Notebook 03. `int("warm")` bricht mit einem `ValueError` ab. Prüfe im Guard deshalb zuerst mit `value.isdigit()` (Notebook 05), ob nur Ziffern im Text stehen.

- Gib die Uhrzeit des ersten und des letzten Fehlers aus (über `[0]` und `[-1]`).
- Zähle die Level aller Zeilen zuerst selbst mit einem Dictionary `counts`: Level als Schlüssel, Anzahl als Wert. Erhöhe den Zähler entweder mit `if level in counts: ... else: ...` oder kürzer mit `counts.get(level, 0) + 1`.
- Zähle die Level dann mit `collections.Counter` und vergleiche mit deiner Zählung: Liefern beide dasselbe? Welche Fassung nimmst du, und warum? Hinweis: `from collections import Counter`, dann `Counter(levels)` für eine Liste `levels` und `.most_common()`.

Erwartet: 3 Fehler (Zeilen 2, 6 und 9), der erste um 08:15:02, der letzte um 12:01:59; 4 × `INFO`, 3 × `ERROR`, 2 × `WARNING`.

- **Untersuchen:** Für das Level allein reicht auch `line.split()[2]`, ganz ohne `maxsplit`. Warum klappt das, das Entpacken in vier Namen ohne `maxsplit` aber nicht?

```python
log = """2026-11-02 08:00:01 INFO Server gestartet
2026-11-02 08:15:02 ERROR Datenbank nicht erreichbar
2026-11-02 08:15:30 INFO Neuer Versuch
2026-11-02 08:16:00 WARNING Antwortzeit 2,3 s
2026-11-02 09:02:11 INFO Anmeldung ada
2026-11-02 09:40:45 ERROR Zeitüberschreitung bei Zahlung
2026-11-02 10:05:00 INFO Bericht erstellt
2026-11-02 11:20:13 WARNING Speicher zu 85 % belegt
2026-11-02 12:01:59 ERROR Datei bericht.pdf fehlt"""

lines = log.splitlines()
print(len(lines), "Zeilen")
```

```python
from collections import Counter

errors = []
for number, line in enumerate(lines, start=1):
    date, time, level, message = line.split(maxsplit=3)  # wie in 05, Übung 5
    if level == "ERROR":
        errors.append((number, time, message))

print(f"Bericht: {len(lines)} Zeilen, davon {len(errors)} Fehler")
for number, time, message in errors:
    print(f"  Zeile {number}, {time}: {message}")
print(f"Erster Fehler um {errors[0][1]}, letzter um {errors[-1][1]}")

# Fassung 1: selbst zählen mit if/else
counts = {}
for line in lines:
    date, time, level, message = line.split(maxsplit=3)
    if level in counts:
        counts[level] += 1
    else:
        counts[level] = 1
print(counts)  # {'INFO': 4, 'ERROR': 3, 'WARNING': 2}

# Fassung 2: selbst zählen mit get, ohne if/else
counts = {}
for line in lines:
    date, time, level, message = line.split(maxsplit=3)
    counts[level] = counts.get(level, 0) + 1

# Fassung 3: Counter zählt in einem Schritt, most_common() sortiert nach Häufigkeit
levels = [line.split()[2] for line in lines]
level_counts = Counter(levels)
print(level_counts.most_common())
print(counts == level_counts)  # True

# Wahl: Die eigene Schleife zeigt, was Counter intern tut. Im Alltag ist Counter
# kürzer und bringt most_common() gleich mit.

# Untersuchen: split() ohne maxsplit zerlegt auch die Meldung in Wörter, aber
# Datum, Uhrzeit und Level stehen trotzdem an Index 0, 1 und 2. Nur das Entpacken
# braucht genau vier Teile; ohne maxsplit=3 sind es mehr (ValueError).
print(lines[1].split())
```

**Übung 9 (Mehrere Wege vergleichen, leicht, für Schnelle):** Für die Schritte aus Übung 1 gibt es jeweils mehrere Wege. Starte jeweils mit `["Milch", "Brot"]`, probiere pro Schritt mindestens zwei Wege aus und entscheide, welchen du nimmst, und warum:

- Mehrere ergänzen: `extend`, `+=` oder `shopping + [...]`.
- Vorne einfügen: `insert(0, ...)` oder `["Kaffee"] + shopping`.
- Abhaken: `remove` (nach Wert), `del shopping[i]` (nach Position) oder eine Comprehension, die alle Vorkommen herausfiltert.

- **Untersuchen:** Welche Wege ändern die bestehende Liste, welche erzeugen eine neue? Was ist bei `["Kaffee", "Milch", "Brot", "Brot"]` der Unterschied zwischen `remove("Brot")` und der Comprehension?

```python
new_items = ["Eier", "Mehl"]

# Mehrere ergänzen: drei Wege mit demselben Ergebnis
a = ["Milch", "Brot"]
a.extend(new_items)  # ändert die bestehende Liste
b = ["Milch", "Brot"]
b += new_items  # ändert die bestehende Liste, wie extend
c = ["Milch", "Brot"] + new_items  # erzeugt eine neue Liste
print(a, b, c)

# Vorne einfügen: zwei Wege
d = ["Milch", "Brot"]
d.insert(0, "Kaffee")  # ändert die Liste
e = ["Kaffee"] + ["Milch", "Brot"]  # neue Liste
print(d, e)

# Abhaken: drei Wege
f = ["Kaffee", "Milch", "Brot", "Brot"]
f.remove("Brot")  # nach Wert, nur das erste Vorkommen
g = ["Kaffee", "Milch", "Brot", "Brot"]
del g[2]  # nach Position
h = [item for item in ["Kaffee", "Milch", "Brot", "Brot"] if item != "Brot"]
print(f, g, h)  # die Comprehension entfernt alle Vorkommen und baut eine neue Liste

# Wahl: extend, insert und remove sagen am klarsten, was passiert, und ändern die
# Liste selbst. + und die Comprehension bauen eine neue Liste; das ist richtig,
# wenn das Original unverändert bleiben soll.
```

## Zusammenfassung

- `list`: geordnet, veränderbar: `append`, `extend`, `insert`, `remove`, `pop`; `+` baut eine neue Liste; `in` und `not in` prüfen, ob ein Wert enthalten ist.
- `.sort()` ändert die Liste und liefert `None`, `sorted()` liefert eine neue Liste; `key=str.lower` sortiert ohne Rücksicht auf Groß- und Kleinschreibung.
- `b = a` kopiert nicht. Eine Kopie legst du mit `a.copy()` an. Werte vergleichst du mit `==`, `is` nur bei `None`.
- `match` mit Mustern für Listen: `case ["move", x, y]:` passt auf genau drei Wörter und nimmt zwei davon auf; `as` gibt einem Teilmuster einen Namen.
- `tuple`: unveränderlich, ideal zum Entpacken: `x, y = point`.
- `dict`: `d[key]`, `d.get(key, default)`, `key in d`, `d.keys()`, `d.values()`, `d.items()`. Unveränderlich: `frozendict`.
- `set`: keine Duplikate, Mengenoperationen `&`, `|`, `-`; leeres Set mit `set()`, denn `{}` ist ein leeres Dictionary.
- `sorted(daten, key=..., reverse=True)` sortiert nach eigenem Kriterium; `key` bekommt eine Funktion ohne Klammern.
- Comprehensions für Listen, Dictionaries und Sets; `[*x for x in listen]` macht flach.
- Aus den Übungen: `collections.Counter` zählt gleiche Einträge, `.most_common()` sortiert nach Häufigkeit.
