| Typ | Schreibweise | geordnet | veränderbar | Duplikate | typischer Einsatz |
| --- | --- | --- | --- | --- | --- |
| `list` | `[1, 2, 3]` | ja | ja | ja | Sammlung gleichartiger Dinge |
| `tuple` | `(1, 2, 3)` | ja | nein | ja | fester Datensatz, z. B. Koordinate |

## Listen

```python
members = ["Michael", "Tim", "Werner"]
print(members[0], members[-1], len(members))
```

Listen sind veränderbar:

```python
members.append("Andreas")             # ein Element anhängen
members.extend(["Lili", "Jana"])      # mehrere anhängen
members.insert(0, "Natalija")         # an Position einfügen
members.remove("Tim")                 # nach Wert entfernen
last = members.pop()                  # letztes entfernen und zurückgeben
print(members, last)
```

```python
print("Jana" in members)
print(members[1:3])  # Slicing wie bei Strings
```

### Sortieren

`sorted()` liefert eine **neue** sortierte Liste, `.sort()` sortiert die Liste **selbst**.

```python
numbers = [5, 2, 9, 1]
print(sorted(numbers))
print(sorted(numbers, reverse=True))
print(numbers)  # unverändert

numbers.sort()
print(numbers)  # jetzt sortiert
```

Mit `key=` bestimmst du, **wonach** sortiert wird. Hier nach der Länge der Namen:

```python
names = ["Tim", "Andreas", "Lili", "Michael"]
sorted(names, key=len)
```

Für eigene Kriterien schreibt man oft eine kleine **Lambda-Funktion** direkt an Ort und Stelle. Hier: nach dem letzten Buchstaben.

```python
sorted(names, key=lambda name: name[-1])
```

> `lambda x: ausdruck` ist eine Funktion ohne Namen. Mehr brauchst du davon für den Anfang nicht. Für alles Längere schreibst du eine normale Funktion (Notebook 06).

## Tupel

Tupel sind wie Listen, aber **unveränderlich**. Sie eignen sich für feste Datensätze.

```python
point = (3, 7)
x, y = point  # Entpacken
print(x, y)
```

```python
point[0] = 5
```

## Dictionaries

Ein Dictionary ordnet **Schlüssel** zu **Werten** zu, wie ein Wörterbuch.

```python
capitals = {"CH": "Bern", "DE": "Berlin", "GB": "London"}
print(capitals["DE"])

capitals["FR"] = "Paris"   # hinzufügen oder ändern
del capitals["GB"]         # entfernen
print(capitals)
```

Fehlende Schlüssel führen zu einem `KeyError`. `.get()` liefert stattdessen einen Standardwert:

```python
capitals["US"]
```

```python
capitals.get("US", "unbekannt")
```

Über ein Dictionary iterieren:

```python
for code, city in capitals.items():
    print(f"{code}: {city}")
```

Dictionaries beschreiben oft einen Datensatz; eine Liste davon ist eine kleine Tabelle:

```python
people = [
    {"name": "Michael", "city": "Zürich", "age": 55},
    {"name": "Tim", "city": "Kiel", "age": 34},
]
for person in people:
    print(f"{person['name']} wohnt in {person['city']}")
```

## Sets

Ein Set enthält jeden Wert nur einmal.

```python
cities = ["Kiel", "Hamburg", "Zürich", "Kiel", "Hamburg"]
unique_cities = set(cities)
unique_cities
```

```python
python_fans = {"Ada", "Tim", "Lili"}
java_fans = {"Tim", "Werner"}
print(python_fans & java_fans)  # Schnittmenge
print(python_fans | java_fans)  # Vereinigung
print(python_fans - java_fans)  # Differenz
```

## Listen kompakt erzeugen

Eine **List Comprehension** baut eine neue Liste aus einer bestehenden:

```python
numbers = [1, 2, 3, 4, 5, 6]
squares = [n ** 2 for n in numbers]
even = [n for n in numbers if n % 2 == 0]
print(squares, even)
```

Das ist dasselbe wie:

```python
squares = []
for n in numbers:
    squares.append(n ** 2)
squares
```

**Übung 1 (Farben):** Gib aus `colors` alle Farben aus, deren Name eine ungerade Länge hat, zusammen mit der Länge.

```python
colors = ["Red", "Green", "Blue", "Purple", "Yellow"]
```

```python
for color in colors:
    if len(color) % 2 != 0:
        print(f"{color} hat Länge {len(color)}")
```

**Übung 2 (Namen zählen):** Zähle mit einem Dictionary, wie oft jeder Name vorkommt.

```python
names = ["Tim", "Tom", "Mike", "Jim", "Tim", "Mike", "James", "Mike"]
```

```python
name_counts = {}
for name in names:
    name_counts[name] = name_counts.get(name, 0) + 1
print(name_counts)
```

**Übung 3 (Duplikate entfernen):** Entferne doppelte Buchstaben aus `"ABCABCDDEFFEGHHH"`, behalte aber die **Reihenfolge** des ersten Auftretens bei: `"ABCDEFGH"`. Tipp: Ein Set merkt sich, was schon da war.

```python
text = "ABCABCDDEFFEGHHH"
seen = set()
result = ""
for letter in text:
    if letter not in seen:
        seen.add(letter)
        result += letter
print(result)
```

**Übung 4 (Bestenliste):** Sortiere die Spieler absteigend nach Punkten und gib die Top 3 aus.

```python
scores = [("Peter", 985), ("Matze", 1000), ("Lili", 1200), ("Tim", 300)]
```

```python
ranking = sorted(scores, key=lambda entry: entry[1], reverse=True)
for place, (name, points) in enumerate(ranking[:3], start=1):
    print(f"{place}. {name}: {points}")
```

## Zusammenfassung

- `list`: geordnet, veränderbar: `append`, `extend`, `remove`, `pop`.
- `tuple`: unveränderlich, ideal zum Entpacken: `x, y = point`.
- `dict`: `d[key]`, `d.get(key, default)`, `d.items()`.
- `set`: keine Duplikate, Mengenoperationen `&`, `|`, `-`.
- `sorted(daten, key=..., reverse=True)` sortiert nach eigenem Kriterium.
