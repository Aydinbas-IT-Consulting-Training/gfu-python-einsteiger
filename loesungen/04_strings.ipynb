## Strings anlegen

Einfache und doppelte Anführungszeichen sind gleichwertig. Drei Anführungszeichen erlauben mehrzeiligen Text.

````python
greeting = "Hallo"
quote = 'Sie sagte: "Python ist toll"'
poem = """Erste Zeile
Zweite Zeile"""
print(quote)
print(poem)
````

`len()` liefert die Länge, `+` verbindet, `*` wiederholt:

````python
print(len(greeting))
print(greeting + " Welt")
print("-" * 20)
````

## Index und Slicing

Jedes Zeichen hat eine Position, beginnend bei **0**. Negative Indizes zählen von hinten.

```
 P  y  t  h  o  n
 0  1  2  3  4  5
-6 -5 -4 -3 -2 -1
```

````python
word = "Python"
print(word[0], word[-1])
````

Mit `[start:stop:step]` schneidest du Teile heraus. `stop` ist wie bei `range()` ausgeschlossen.

````python
print(word[0:2])   # "Py"
print(word[2:])    # ab Index 2 bis Ende
print(word[:-2])   # alles außer den letzten zwei
print(word[::-1])  # umgedreht
````

## Strings sind unveränderlich

Einzelne Zeichen kann man nicht ersetzen. Methoden liefern stattdessen einen **neuen** String.

````python
word[0] = "J"
````

````python
new_word = "J" + word[1:]
new_word
````

## Wichtige Methoden

````python
text = "  Einstieg in Python  "
print(repr(text.strip()))         # Leerzeichen am Rand entfernen
print(text.upper())
print(text.lower())
print(text.replace("Python", "Java"))
print(text.strip().startswith("Ein"))
print("Python" in text)           # Enthalten-Test
print(text.count("i"))
print(text.find("Python"))        # Position oder -1
````

Prüfmethoden helfen bei Eingaben:

````python
print("2026".isdigit(), "abc".isalpha(), "Abc".istitle())
````

## Zerlegen und Verbinden

`split()` zerlegt einen String in eine Liste, `join()` fügt eine Liste zusammen.

````python
line = "Ada,Lovelace,1815"
parts = line.split(",")
print(parts)

words = ["Python", "ist", "toll"]
print(" ".join(words))
````

## Über Zeichen iterieren

````python
for letter in "Hallo":
    print(letter.upper(), end=".")
````

## Übungen

**Übung 1 (Vokale ersetzen):** Schreibe Code, der in `text` alle Vokale (inklusive Umlaute) durch `_` ersetzt. Aus `"Es gibt viel zu entdecken!"` wird `"_s g_bt v__l z_ _ntd_ck_n!"`.

````python
text = "Es gibt viel zu entdecken!"
vowels = "AEIOUÄÖÜaeiouäöü"

result = ""
for letter in text:
    if letter in vowels:
        result += "_"
    else:
        result += letter
print(result)
````

**Übung 2 (Zeichen wiederholen):** Jedes Zeichen soll so oft wiederholt werden wie seine Position (ab 1). Aus `"ABCD"` wird `"ABBCCCDDDD"`. Tipp: `enumerate()`.

````python
text = "ABCD"
result = ""
for position, letter in enumerate(text, start=1):
    result += letter * position
print(result)
````

**Übung 3 (Palindrom):** Prüfe, ob ein Wort vorwärts und rückwärts gleich ist, ohne Groß- und Kleinschreibung zu beachten. Teste mit `"Rentner"` und `"Python"`.

````python
word = "Rentner"
normalized = word.lower()
print(f"{word} ist Palindrom: {normalized == normalized[::-1]}")
````

**Übung 4 (Initialen):** Erzeuge aus `"Ada King Lovelace"` die Initialen `"A. K. L."`.

````python
full_name = "Ada King Lovelace"
initials = []
for part in full_name.split():
    initials.append(part[0] + ".")
print(" ".join(initials))
````

## Zusammenfassung

- `in` prüft, ob ein Teilstring enthalten ist.
- `split()` zerlegt, `"trenner".join(liste)` verbindet.
