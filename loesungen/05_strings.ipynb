## Strings anlegen

Einfache und doppelte Anführungszeichen sind gleichwertig. Drei Anführungszeichen erlauben mehrzeiligen Text.

````python
greeting = "Hallo"
quote = 'Sie sagte: "Python ist toll"'
poem = """Erste Zeile
Zweite Zeile"""
print(quote)
print(poem)
````

Manche Zeichen schreibst du mit einem Backslash (**Escape-Sequenz**): `\n` beginnt eine neue Zeile, `\t` setzt einen Tabulator, `\\` steht für den Backslash selbst.

````python
print("Name:\tAda\nOrt:\tKiel")
print("C:\\Users\\ada")
````

`len()` liefert die Länge, `+` verbindet, `*` wiederholt:

````python
print(len(greeting))
print(greeting + " Welt")
print("-" * 20)
````

## Index und Slicing

Jedes Zeichen hat eine Position, beginnend bei **0**. Negative Indizes zählen von hinten.

```
 P  y  t  h  o  n
 0  1  2  3  4  5
-6 -5 -4 -3 -2 -1
```

````python
word = "Python"
print(word[0], word[-1])
````

Mit `[start:stop:step]` schneidest du Teile heraus. `stop` ist wie bei `range()` ausgeschlossen.

````python
print(word[0:2])   # "Py"
print(word[2:])    # ab Index 2 bis Ende
print(word[:-2])   # alles außer den letzten zwei
print(word[::-1])  # umgedreht
````

**Kurzübung 2 (Rechnungsnummer zerlegen):** Gegeben ist `invoice = "RE-2026-0042"`. Welche Ausgabe erzeugen `invoice[0]`, `invoice[-1]`, `invoice[3:7]` und `invoice[-4:]`? Schreibe deine Vorhersage mit Begründung auf und prüfe sie dann. Gib anschließend `Jahr 2026, Nr. 0042` aus, ohne die Zahlen abzutippen.

````python
invoice = "RE-2026-0042"
print(invoice[0])    # "R": Index 0 ist das erste Zeichen
print(invoice[-1])   # "2": -1 ist das letzte Zeichen
print(invoice[3:7])  # "2026": Index 3 bis 6, 7 ist ausgeschlossen
print(invoice[-4:])  # "0042": die letzten vier Zeichen

print(f"Jahr {invoice[3:7]}, Nr. {invoice[-4:]}")
````

**Kurzübung 4 (Namen im Adressbuch finden):** Im Adressbuch steht `"Müller"`, ins Suchfeld wurde `"  MÜLLER "` eingegeben. Vergleiche erst mit `==`, dann beide Seiten mit `strip().lower()`. Welche Variante findet den Eintrag, und warum?

- **Untersuchen:** Im Adressbuch steht `"Hauptstraße"`, gesucht wird `"HAUPTSTRASSE"`. Reicht `lower()` hier auch? Probiere `casefold()` aus und erkläre den Unterschied.

````python
stored, searched = "Müller", "  MÜLLER "
print(stored == searched)  # False: Leerzeichen und Großschreibung
print(stored.strip().lower() == searched.strip().lower())  # True

# Untersuchen: lower() lässt ß stehen, casefold() macht daraus ss.
street, query = "Hauptstraße", "HAUPTSTRASSE"
print(street.lower() == query.lower())  # False: "hauptstraße" gegen "hauptstrasse"
print(street.casefold() == query.casefold())  # True
````

## Zerlegen und Verbinden

`split()` zerlegt einen String in eine Liste, `join()` fügt eine Liste zusammen. `splitlines()` zerlegt mehrzeiligen Text in seine Zeilen.

````python
line = "Ada,Lovelace,1815"
parts = line.split(",")
print(parts)

words = ["Python", "ist", "toll"]
print(" ".join(words))

notes = "Milch kaufen\nBrot kaufen\nZahnarzt anrufen"
print(notes.splitlines())
````

**Kurzübung 5 (Datum umformatieren):** Ein Export liefert das Datum als `"2026-11-02"`. Zerlege es mit `split("-")` in Jahr, Monat und Tag und gib es als `02.11.2026` aus: einmal mit einem f-String, einmal mit `".".join(...)`.

````python
iso_date = "2026-11-02"
year, month, day = iso_date.split("-")

print(f"{day}.{month}.{year}")      # Fassung 1: f-String
print(".".join([day, month, year]))  # Fassung 2: join

# Beide liefern 02.11.2026. Der f-String zeigt das Format auf einen Blick,
# join lohnt sich, wenn schon eine Liste von Teilen vorliegt.
````

**Kurzübung 6 (Laute Betreffzeile):** Ein Mailfilter soll Betreffzeilen mit vielen Großbuchstaben erkennen. Zähle mit einer Schleife über die Zeichen von `"JETZT GRATIS testen!!!"`, wie viele Großbuchstaben darin stehen, und gib `11 Großbuchstaben` aus. Tipp: `char.isupper()`.

- **Untersuchen:** Warum zählen Leerzeichen und `!` nicht mit?

````python
subject = "JETZT GRATIS testen!!!"
upper_count = 0
for char in subject:
    if char.isupper():
        upper_count += 1
print(f"{upper_count} Großbuchstaben")

# Untersuchen: Leerzeichen und "!" haben keine Groß- oder Kleinform,
# deshalb liefert isupper() für sie False.
````

**Übung 1 (IBAN maskieren, leicht):** Für einen Kontoauszug soll aus `"DE89 3704 0044 0532 0130 00"` die Anzeige `******************3000` werden: Leerzeichen entfernen, nur die letzten 4 Zeichen zeigen.

Entferne die Leerzeichen auf zwei Arten: zuerst mit einer `for`-Schleife, die jedes Zeichen außer `" "` an einen neuen String anhängt, dann mit `replace(" ", "")`. Welche Fassung nimmst du, und warum?

````python
iban = "DE89 3704 0044 0532 0130 00"

# Fassung 1: selbst geschrieben, Zeichen für Zeichen
compact = ""
for char in iban:
    if char != " ":
        compact += char
print("*" * (len(compact) - 4) + compact[-4:])

# Fassung 2: mit replace
compact = iban.replace(" ", "")
print("*" * (len(compact) - 4) + compact[-4:])

# Vergleich: Fassung 1 zeigt, was replace intern tut; Fassung 2 ist kürzer und
# sagt direkt, was gemeint ist. Im Alltag nimmt man replace.
````

**Übung 2 (E-Mail-Adresse aus dem Formular, leicht):** Ein Anmeldeformular liefert `"  Max.Mustermann@Firma.DE "`. Bereinige die Eingabe mit `strip()` und `lower()`, prüfe mit `in`, ob ein `@` enthalten ist, und zerlege die Adresse mit `split("@")` in Benutzername und Domain. Erwartet: `Benutzer: max.mustermann, Domain: firma.de`; fehlt das `@`, meldet dein Code `Ungültige Adresse: ...`.

- **Untersuchen:** Was passiert beim Zerlegen, wenn das `@` fehlt, etwa bei `"max.mustermann.firma.de"`? Warum lohnt sich die Prüfung mit `in` vorher?

````python
raw = "  Max.Mustermann@Firma.DE "
email = raw.strip().lower()
if "@" in email:
    user, domain = email.split("@")
    print(f"Benutzer: {user}, Domain: {domain}")
else:
    print(f"Ungültige Adresse: {email}")

# Untersuchen: Ohne @ liefert split("@") nur einen Teil; das Entpacken in zwei
# Namen wirft ValueError: not enough values to unpack. Die Prüfung mit in fängt
# das vorher ab und gibt eine verständliche Meldung aus.
````

**Übung 3 (Initialen aus Formulareingabe, leicht):** Erzeuge aus der Formulareingabe `"  ada  king lovelace "` die Initialen `"A. K. L."`. Zerlege den Namen mit `split()` und baue die Initialen auf zwei Arten:

- mit einer `for`-Schleife, die jede Initiale samt Punkt mit `+=` an einen String anhängt,
- mit `join()`: Sammle die Initialen dazu in einer Liste und verbinde sie danach mit `" "`. Eine leere Liste legst du mit `letters = []` an, `letters.append(x)` hängt `x` hinten an (mehr dazu in Notebook 06).

Welche Fassung nimmst du, und warum? Probiere außerdem `split(" ")` statt `split()`: Warum bricht der Code dann ab?

````python
full_name = "  ada  king lovelace "

# Fassung 1: Schleife mit +=
initials = ""
for part in full_name.split():
    initials += part[0].upper() + ". "
print(initials.strip())  # strip entfernt das Leerzeichen am Ende

# Fassung 2: Liste sammeln, mit join verbinden
letters = []
for part in full_name.split():
    letters.append(part[0].upper() + ".")
print(" ".join(letters))

# Vergleich: join setzt den Trenner nur zwischen die Teile, das strip am Ende
# entfällt. split(" ") liefert für doppelte Leerzeichen leere Teile "",
# und ""[0] wirft IndexError; split() ohne Argument überspringt sie.
````

**Übung 4 (Passwortprüfung, mittel):** Ein Passwort ist sicher, wenn es mindestens 12 Zeichen, eine Ziffer, einen Groß- und einen Kleinbuchstaben hat. Der Startcode prüft Länge und Ziffer. Sage vorher, was er für `"Sommer2026"` und `"sommer"` ausgibt, führe den Code dann aus und ergänze die Prüfung auf Groß- und Kleinbuchstaben.

- **Untersuchen:** Was ändert sich, wenn in der Schleife ein `else: has_digit = False` steht? Warum stehen die Meldungen nach der Schleife?
- **Ändern:** Ergänze Merker für Groß- und Kleinbuchstaben.

````python
password = "Sommer2026"
has_digit = False
for char in password:
    if char.isdigit():
        has_digit = True
if len(password) < 12:
    print("kürzer als 12 Zeichen")
if not has_digit:
    print("Ziffer fehlt")
````

````python
password = "Sommer2026"
has_digit = False
has_upper = False
has_lower = False
for char in password:
    if char.isdigit():
        has_digit = True
    elif char.isupper():
        has_upper = True
    elif char.islower():
        has_lower = True
if len(password) < 12:
    print("kürzer als 12 Zeichen")
if not has_digit:
    print("Ziffer fehlt")
if not has_upper:
    print("Großbuchstabe fehlt")
if not has_lower:
    print("Kleinbuchstabe fehlt")

# Vorhersage: "Sommer2026" -> kürzer als 12 Zeichen;
# "sommer" -> kürzer als 12 Zeichen, Ziffer fehlt.
# Untersuchen: Mit else: has_digit = False zählt nur das letzte Zeichen.
# In der Schleife käme die Meldung einmal pro Zeichen statt einmal pro Passwort.
````

**Übung 6 (Fotonamen vereinheitlichen, schwer):** Dein Handy speichert Fotos als `IMG_20261030_081502.jpg`. Der Startcode macht aus einem Namen das Datum. Sage die Ausgabe vorher, führe den Code dann aus und erweitere ihn: Für jeden Namen in `photos` entsteht `2026-10-30_08-15-02.jpg` (Endung klein); Namen, die nicht mit `IMG_` beginnen, bleiben unverändert und werden gemeldet. Tipp: `removeprefix()` und `rpartition()` (siehe Python-Dokumentation zu String-Methoden) sparen dir Zählarbeit.

- **Untersuchen:** Was liefert der Startcode für `"Scan Mietvertrag.pdf"`? Woher kommen die Zahlen 4 und 12?
- **Ändern:** Ergänze Uhrzeit und Endung, durchlaufe die Liste und überspringe fremde Namen.

````python
photos = ["IMG_20261030_081502.jpg", "IMG_20261101_174233.JPG", "Scan Mietvertrag.pdf"]

name = "IMG_20261030_081502.jpg"
day = name[4:12]
print(f"{day[:4]}-{day[4:6]}-{day[6:]}")
````

````python
photos = ["IMG_20261030_081502.jpg", "IMG_20261101_174233.JPG", "Scan Mietvertrag.pdf"]
for name in photos:
    if not name.startswith("IMG_"):
        print(f"übersprungen: {name}")
        continue
    stem, _, suffix = name.removeprefix("IMG_").rpartition(".")
    day, clock = stem.split("_")
    date_part = f"{day[:4]}-{day[4:6]}-{day[6:]}"
    time_part = f"{clock[:2]}-{clock[2:4]}-{clock[4:]}"
    new_name = f"{date_part}_{time_part}.{suffix.lower()}"
    print(f"{name} -> {new_name}")

# Untersuchen: Für "Scan Mietvertrag.pdf" liefert der Startcode " Mie-tv-er",
# denn Slicing prüft nichts. 4 ist die Länge von "IMG_", 12 = 4 + 8 Ziffern.
````

## Zusammenfassung

- Index ab 0, negativ von hinten; Slicing mit `[start:stop:step]`.
- Escape-Sequenzen: `\n` neue Zeile, `\t` Tabulator, `\\` Backslash.
- Strings sind unveränderlich; Methoden liefern neue Strings.
- Wichtige Methoden: `strip`, `upper`, `lower`, `replace`, `startswith`, `find`, `count`.
- `in` prüft, ob ein Teilstring enthalten ist.
- `split()` zerlegt, `"trenner".join(liste)` verbindet; `splitlines()` zerlegt mehrzeiligen Text in Zeilen.
- Aus den Übungen: `casefold()` vergleicht ohne Groß- und Kleinschreibung gründlicher als `lower()` (aus `ß` wird `ss`); `removeprefix()` entfernt einen festen Anfang, `rpartition(".")` trennt am letzten Punkt, etwa die Dateiendung.
