## Strings anlegen

Einfache und doppelte Anführungszeichen sind gleichwertig. Drei Anführungszeichen erlauben mehrzeiligen Text.

````python
greeting = "Hallo"
quote = 'Sie sagte: "Python ist toll"'
poem = """Erste Zeile
Zweite Zeile"""
print(quote)
print(poem)
````

`len()` liefert die Länge, `+` verbindet, `*` wiederholt:

````python
print(len(greeting))
print(greeting + " Welt")
print("-" * 20)
````

## Index und Slicing

Jedes Zeichen hat eine Position, beginnend bei **0**. Negative Indizes zählen von hinten.

```
 P  y  t  h  o  n
 0  1  2  3  4  5
-6 -5 -4 -3 -2 -1
```

````python
word = "Python"
print(word[0], word[-1])
````

Mit `[start:stop:step]` schneidest du Teile heraus. `stop` ist wie bei `range()` ausgeschlossen.

````python
print(word[0:2])   # "Py"
print(word[2:])    # ab Index 2 bis Ende
print(word[:-2])   # alles außer den letzten zwei
print(word[::-1])  # umgedreht
````

**Übung 1 (IBAN maskieren, leicht):** Für einen Kontoauszug soll aus `"DE89 3704 0044 0532 0130 00"` die Anzeige `******************3000` werden: Leerzeichen entfernen, nur die letzten 4 Zeichen zeigen.

Entferne die Leerzeichen auf zwei Arten: zuerst mit einer `for`-Schleife, die jedes Zeichen außer `" "` an einen neuen String anhängt, dann mit `replace(" ", "")`. Welche Fassung nimmst du, und warum?

````python
iban = "DE89 3704 0044 0532 0130 00"

# Fassung 1: selbst geschrieben, Zeichen für Zeichen
compact = ""
for char in iban:
    if char != " ":
        compact += char
print("*" * (len(compact) - 4) + compact[-4:])

# Fassung 2: mit replace
compact = iban.replace(" ", "")
print("*" * (len(compact) - 4) + compact[-4:])

# Vergleich: Fassung 1 zeigt, was replace intern tut; Fassung 2 ist kürzer und
# sagt direkt, was gemeint ist. Im Alltag nimmt man replace.
````

## Wichtige Methoden

````python
text = "  Einstieg in Python  "
print(repr(text.strip()))         # Leerzeichen am Rand entfernen
print(text.upper())
print(text.lower())
print(text.replace("Python", "Java"))
print(text.strip().startswith("Ein"))
print("Python" in text)           # Enthalten-Test
print(text.count("i"))
print(text.find("Python"))        # Position oder -1
````

Prüfmethoden helfen bei Eingaben:

````python
print("2026".isdigit(), "abc".isalpha(), "Abc".istitle())
````

**Übung 2 (Namen im Adressbuch finden, leicht):** Im Adressbuch steht `"Straße"`, gesucht wird `"  STRASSE"`. Vergleiche erst mit `==`, dann mit `strip().lower()`, dann mit `strip().casefold()`. Welche Variante findet den Eintrag, und warum?

````python
stored, searched = "Straße", "  STRASSE"
print(stored == searched)  # False
print(stored.lower() == searched.strip().lower())  # False: ß bleibt ß
print(stored.casefold() == searched.strip().casefold())  # True: ß wird ss
````

## Zerlegen und Verbinden

`split()` zerlegt einen String in eine Liste, `join()` fügt eine Liste zusammen.

````python
line = "Ada,Lovelace,1815"
parts = line.split(",")
print(parts)

words = ["Python", "ist", "toll"]
print(" ".join(words))
````

**Übung 3 (Initialen aus Formulareingabe, leicht):** Erzeuge aus der Formulareingabe `"  ada  king lovelace "` die Initialen `"A. K. L."`. Verwende zuerst `split(" ")`: Warum bricht der Code dann ab?

````python
full_name = "  ada  king lovelace "
initials = ""
for part in full_name.split():  # split(" ") liefert leere Teile -> IndexError
    initials += part[0].upper() + ". "
print(initials.strip())
````

**Übung 4 (Passwortprüfung, mittel):** Ein Passwort ist sicher, wenn es mindestens 12 Zeichen, eine Ziffer, einen Groß- und einen Kleinbuchstaben hat. Der Startcode prüft Länge und Ziffer. Sage vorher, was er für `"Sommer2026"` und `"sommer"` ausgibt, führe den Code dann aus und ergänze die Prüfung auf Groß- und Kleinbuchstaben.

- Untersuchen: Was ändert sich, wenn in der Schleife ein `else: has_digit = False` steht? Warum stehen die Meldungen nach der Schleife?
- Ändern: Ergänze Merker für Groß- und Kleinbuchstaben.

````python
password = "Sommer2026"
has_digit = False
for char in password:
    if char.isdigit():
        has_digit = True
if len(password) < 12:
    print("zu kurz")
if not has_digit:
    print("Ziffer fehlt")
````

````python
password = "Sommer2026"
has_digit = has_upper = has_lower = False
for char in password:
    if char.isdigit():
        has_digit = True
    elif char.isupper():
        has_upper = True
    elif char.islower():
        has_lower = True
if len(password) < 12:
    print("zu kurz")
if not has_digit:
    print("Ziffer fehlt")
if not has_upper:
    print("Großbuchstabe fehlt")
if not has_lower:
    print("Kleinbuchstabe fehlt")

# Vorhersage: "Sommer2026" -> zu kurz; "sommer" -> zu kurz, Ziffer fehlt.
# Untersuchen: Mit else: has_digit = False zählt nur das letzte Zeichen.
# In der Schleife käme die Meldung einmal pro Zeichen statt einmal pro Passwort.
````

**Übung 6 (Fotonamen vereinheitlichen, schwer):** Dein Handy speichert Fotos als `IMG_20261030_081502.jpg`. Der Startcode macht aus einem Namen das Datum. Sage die Ausgabe vorher, führe den Code dann aus und erweitere ihn: Für jeden Namen in `photos` entsteht `2026-10-30_08-15-02.jpg` (Endung klein); Namen, die nicht mit `IMG_` beginnen, bleiben unverändert und werden gemeldet. Tipp: `removeprefix()` und `rpartition()` (siehe Python-Dokumentation zu String-Methoden) sparen dir Zählarbeit.

- Untersuchen: Was liefert der Startcode für `"Scan Mietvertrag.pdf"`? Woher kommen die Zahlen 4 und 12?
- Ändern: Ergänze Uhrzeit und Endung, durchlaufe die Liste und überspringe fremde Namen.

````python
photos = ["IMG_20261030_081502.jpg", "IMG_20261101_174233.JPG", "Scan Mietvertrag.pdf"]

name = "IMG_20261030_081502.jpg"
day = name[4:12]
print(f"{day[:4]}-{day[4:6]}-{day[6:]}")
````

````python
photos = ["IMG_20261030_081502.jpg", "IMG_20261101_174233.JPG", "Scan Mietvertrag.pdf"]
for name in photos:
    if not name.startswith("IMG_"):
        print(f"übersprungen: {name}")
        continue
    stem, _, suffix = name.removeprefix("IMG_").rpartition(".")
    day, clock = stem.split("_")
    date_part = f"{day[:4]}-{day[4:6]}-{day[6:]}"
    time_part = f"{clock[:2]}-{clock[2:4]}-{clock[4:]}"
    new_name = f"{date_part}_{time_part}.{suffix.lower()}"
    print(f"{name} -> {new_name}")

# Untersuchen: Für "Scan Mietvertrag.pdf" liefert der Startcode " Mie-tv-er",
# denn Slicing prüft nichts. 4 ist die Länge von "IMG_", 12 = 4 + 8 Ziffern.
````

## Zusammenfassung

- Index ab 0, negativ von hinten; Slicing mit `[start:stop:step]`.
- Strings sind unveränderlich; Methoden liefern neue Strings.
- Wichtige Methoden: `strip`, `upper`, `lower`, `replace`, `startswith`, `find`, `count`.
- Für Vergleiche ohne Groß- und Kleinschreibung `casefold()` statt `lower()` (aus `ß` wird `ss`).
- `in` prüft, ob ein Teilstring enthalten ist.
- `split()` zerlegt, `"trenner".join(liste)` verbindet; `splitlines()` zerlegt mehrzeiligen Text in Zeilen.
- `removeprefix()` entfernt einen festen Anfang, `rpartition(".")` trennt am letzten Punkt, etwa die Dateiendung.
