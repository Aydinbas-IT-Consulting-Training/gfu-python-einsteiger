## Die `for`-Schleife

`for` läuft über alle Elemente einer Folge, etwa die Zeichen eines Strings oder die Einträge einer Liste:

````python
for letter in "Python":
    print(letter)
````

````python
for city in ["Köln", "Berlin", "München"]:
    print(f"Schulung in {city}")
````

## Index und Wert mit `enumerate()`

Brauchst du zusätzlich die Position, nimmst du `enumerate()` statt `range(len(...))`. `enumerate()` liefert in jeder Runde ein Paar aus Position und Wert; `for position, name in ...` entpackt es in zwei Namen (Entpacken, Notebook 01):

## Die `while`-Schleife

`while` wiederholt, **solange** eine Bedingung wahr ist. Sinnvoll, wenn die Anzahl der Durchläufe vorher nicht feststeht.

![Flussdiagramm der while-Schleife: balance < 2000 wird vor jedem Durchlauf geprüft; die Tabelle zeigt, dass die Bedingung bei years = 15 und balance = 2078.93 zum ersten Mal falsch ist.](../assets/04_while_schleife.png)

````python
balance = 1000.0
years = 0
while balance < 2000:
    balance *= 1.05
    years += 1
print(f"Nach {years} Jahren hat sich das Guthaben verdoppelt: {balance:.2f} €")
````

> **Achtung:** Ändert sich die Bedingung nie, läuft die Schleife endlos. Abbrechen im Notebook mit dem Stopp-Button (■), im Terminal mit `Strg + C`.

## `break` und `continue`

- `break` beendet die Schleife sofort.
- `continue` springt zum nächsten Durchlauf.

![Sprungziele: break springt zur ersten Zeile hinter der Schleife, continue springt zurück zum Schleifenkopf for number in range(1, 11), die Ausgabe ist 1 2 4 5 7 8 10.](../assets/04_break_continue.png)

````python
import random

tries = 0
while True:
    roll = random.randint(1, 6)
    tries += 1
    if roll == 6:
        break
print(f"Nach {tries} Würfen eine 6")
````

````python
for number in range(1, 11):
    if number % 3 == 0:
        continue
    print(number, end=" ")
````

**Kurzübung 8 (Bestellungen abarbeiten):** Ein Export aus dem Webshop liefert `orders = ["A-101", "A-102", "", "A-104", "STOP", "A-106"]`. Gib für jede Bestellung `Bearbeite A-101` usw. aus. Leere Einträge überspringst du mit `continue`, bei `"STOP"` endet die Verarbeitung mit `break`. Welche Bestellnummern erscheinen? Schreibe deine Vorhersage auf, bevor du den Code ausführst.

````python
orders = ["A-101", "A-102", "", "A-104", "STOP", "A-106"]
for order in orders:
    if order == "":
        continue  # leere Zeile: nächste Bestellung
    if order == "STOP":
        break  # Ende des Exports: Schleife verlassen
    print(f"Bearbeite {order}")

# Ausgabe: A-101, A-102, A-104. A-106 steht hinter STOP und wird nie erreicht.
````

**Übung 3 (Sparplan und Ratenkauf, leicht):** Zwei Rechnungen rund ums Geld:

- Du legst 1000 € zu 3 % Zinsen pro Jahr an; die Zinsen bleiben auf dem Konto. Gib mit einer `for`-Schleife über `range` den Kontostand nach jedem der ersten 10 Jahre aus, mit 2 Nachkommastellen, etwa `Jahr 1: 1030.00 €`.
- Du kaufst einen Laptop für 1199 € auf Raten und zahlst jeden Monat 150 € ab; die letzte Rate ist nur noch der Rest. Zähle mit einer `while`-Schleife, wie viele Raten es werden. Erwartet: `8 Raten, letzte Rate 149 €`.

Warum passt `for` zum ersten Teil und `while` zum zweiten?

````python
# Teil 1: feste Zahl von Jahren -> for
balance = 1000.0
for year in range(1, 11):
    balance *= 1.03
    print(f"Jahr {year}: {balance:.2f} €")

# Teil 2: Ende hängt vom Restbetrag ab -> while
remaining = 1199
installments = 0
while remaining > 0:
    payment = 150 if remaining >= 150 else remaining  # letzte Rate: nur der Rest
    remaining -= payment
    installments += 1
print(f"{installments} Raten, letzte Rate {payment} €")

# for, wenn die Zahl der Durchläufe vorher feststeht (10 Jahre);
# while, wenn erst die Rechnung zeigt, wann Schluss ist (8 Raten).
````

````python
# Untersuchen: range(10, 60, 20) liefert nur 10, 30 und 50, denn 70 liegt über 60.
# Mit range(10, 80, 20) käme die 70 dazu, aber die Stunde bleibt fest 06:
# Es erschiene 06:70 statt 07:10. Deshalb in Minuten seit Mitternacht rechnen.

for minutes in range(6 * 60 + 10, 9 * 60, 20):  # 370 bis vor 540
    hours = minutes // 60
    mins = minutes % 60
    print(f"{hours:02}:{mins:02}")
# 9 Abfahrten: 06:10, 06:30, 06:50, 07:10, ... 08:50

print("Mit divmod:")
for minutes in range(6 * 60 + 10, 9 * 60, 20):
    hours, mins = divmod(minutes, 60)
    print(f"{hours:02}:{mins:02}")
````

````python
import random

# Untersuchen: Ohne 42 in der Liste gibt der Startcode nur "zu klein"/"zu groß"
# aus, "Richtig!" erscheint nie. Die Variable guessed merkt sich den Treffer:
secret = 42
guessed = False
for guess in [50, 25, 30]:
    if guess < secret:
        print(guess, "ist zu klein")
    elif guess > secret:
        print(guess, "ist zu groß")
    else:
        print("Richtig!")
        guessed = True
        break
if not guessed:
    print("Nicht erraten")
# for passt nicht mehr, wenn ein Mensch tippt: Wie viele Tipps er braucht, steht
# vorher nicht fest. Das Spiel endet erst, wenn der Tipp stimmt -> while.

# Ändern: das echte Spiel
secret = random.randint(1, 100)
tries = 0
while True:
    guess = int(input("Dein Tipp (1-100): "))
    tries += 1
    if guess < secret:
        print("zu klein")
    elif guess > secret:
        print("zu groß")
    else:
        print(f"Richtig nach {tries} Versuchen")
        break
````

**Übung 9 (Der Computer rät, schwer, für Schnelle):** Kehre die Rollen aus Übung 8 um: Du denkst dir eine Zahl von 1 bis 100 (im Code `secret = 42`), der Computer rät. Er tippt immer die Mitte des noch möglichen Bereichs und verkleinert den Bereich nach jedem „zu klein“ oder „zu groß“. Gib jeden Tipp aus und am Ende die Zahl der Versuche. Erwartet für 42: `Richtig nach 7 Versuchen`.

- **Untersuchen:** Wie viele Versuche braucht der Computer höchstens, egal welche Zahl du dir denkst?

````python
# Der Computer rät deine Zahl, indem er den Bereich jedes Mal halbiert.
secret = 42  # deine Zahl
low, high = 1, 100
tries = 0
while True:
    guess = (low + high) // 2
    tries += 1
    if guess < secret:
        print(f"{guess} ist zu klein")
        low = guess + 1
    elif guess > secret:
        print(f"{guess} ist zu groß")
        high = guess - 1
    else:
        print(f"Richtig nach {tries} Versuchen")
        break
# Untersuchen: höchstens 7 Versuche, denn jeder Tipp halbiert den Bereich:
# 100 -> 50 -> 25 -> 12 -> 6 -> 3 -> 1 Kandidaten
````
