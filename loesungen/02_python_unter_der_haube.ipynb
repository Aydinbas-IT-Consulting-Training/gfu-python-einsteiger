## Alles ist ein Objekt

In Java gibt es primitive Typen wie `int` und Objekte. In Python gibt es nur **Objekte**. Jeder Wert hat einen Typ und bringt **Methoden** mit, die man mit einem Punkt aufruft.

Bei Strings ist das vertraut:

````python
"hallo".upper()
````

Aber auch eine Zahl ist ein Objekt mit Methoden:

````python
(42).bit_length()  # wie viele Bits braucht 42?
````

````python
(3.75).as_integer_ratio()  # 3.75 als Bruch
````

> Die Klammern um `42` sind nötig, weil Python `42.` sonst als Kommazahl liest.

Sogar Funktionen und Typen sind Objekte:

````python
print(type(42))
print(type(print))
print(type(int))
````

## Was kann ein Objekt? `dir()` verrät es

`dir()` listet alle Attribute und Methoden eines Objekts auf:

````python
print(dir("hallo"))
````

Die Namen mit doppelten Unterstrichen am Anfang der Liste (`__add__`, `__len__` …) heißen **Dunder-Methoden** (*double underscore*). Sie sind der Schlüssel zum nächsten Abschnitt. Die Namen ohne Unterstriche am Ende (`capitalize`, `upper` …) sind die Methoden, die du selbst aufrufst.

## Operatoren sind Methoden

Wenn du `1 + 2` schreibst, ruft Python in Wahrheit eine Methode des linken Objekts auf:

````python
print(1 + 2)
print((1).__add__(2))
````

Deshalb bedeutet ein Operator je nach Typ etwas anderes. Jeder Typ entscheidet selbst, was er heißt; für Strings heißt `*` wiederholen:

````python
print("ab" * 3, "ab".__mul__(3))
````

Auch viele eingebaute Funktionen fragen nur das Objekt:

| Du schreibst | Python ruft auf |
| --- | --- |
| `a + b` | `a.__add__(b)` |
| `len(x)` | `x.__len__()` |

> In echtem Code schreibst du natürlich `a + b`, nicht `a.__add__(b)`. Eigene Dunder-Methoden schreibst du erst in Notebook 12 (optional).

Ein Typ, der eine Methode nicht hat, kann die Operation nicht. Das erklärt Fehlermeldungen wie diese:

````python
"Alter: " + 18
````

`str.__add__` kann nur Strings verbinden. Mit einer Zahl weiß es nichts anzufangen.

## Variablen haben keinen Typ, Werte schon

Eine Variable ist nur ein **Name**, der auf ein Objekt zeigt. Den Typ trägt das Objekt. Deshalb darf derselbe Name nacheinander auf ganz verschiedene Werte zeigen (**dynamische Typisierung**):

````python
value = 42
print(type(value))
value = "zweiundvierzig"
print(type(value))
````

Praktisch ist das selten eine gute Idee. Ein Name sollte immer dieselbe Art von Wert bedeuten.

Mit `isinstance()` prüfst du, ob ein Wert einen bestimmten Typ hat:

````python
print(isinstance(42, int))
print(isinstance(42, str))
````

## Eingebaute Funktionen sind nur Namen

`print`, `len`, `sum`, `max`, `list`, `str` … sind keine Schlüsselwörter, sondern ganz normale Namen aus dem Modul `builtins`. Du kannst sie deshalb, meist aus Versehen, **überschreiben**:

````python
import builtins

print(len(dir(builtins)), "eingebaute Namen")
````

Ein typischer Anfängerfehler: Eine Variable heißt `list`, `sum` oder `max`. Im Beispiel ist `[3, 1, 2]` eine **Liste**, also mehrere Werte in eckigen Klammern; Listen lernst du in Notebook 06 genauer kennen. `list("abc")` macht aus einem Text eine Liste seiner Zeichen.

````python
list = [3, 1, 2]  # überschreibt den eingebauten Namen list
print(list)
````

````python
letters = list("abc")  # TypeError: list ist jetzt eine Liste, keine Funktion mehr
````

Mit `del` entfernst du deinen eigenen Namen. Danach ist der eingebaute wieder sichtbar:

````python
del list
list("abc")
````

Python sucht einen Namen in dieser Reihenfolge: erst **lokal** (in der Funktion), dann im **Modul** (Notebook oder Datei), zuletzt in den **builtins**. Der erste Treffer gewinnt.

Echte **Schlüsselwörter** wie `if`, `for` oder `def` lassen sich dagegen nicht überschreiben:

````python
import keyword

print(keyword.kwlist)
````

````python
for = 5
````

> Faustregel: Nenne nichts `list`, `dict`, `str`, `sum`, `max`, `min`, `id`, `type`, `input` oder `print`. Besser sind sprechende Namen wie `numbers`, `total` oder `user_input`. Der Linter ruff warnt davor in deinen Skripten, wenn du seine Regel `A001` einschaltest, z. B. `uv run ruff check --select A001 greeting.py`.

**Übung 3 (Werte aus Formular und Kasse prüfen, leicht):** Ein Bestellformular liefert die Menge und den Stückpreis, die Kasse den Rabatt in Euro (siehe Zelle unten). Prüfe zuerst mit `isinstance()`, welche Werte schon Zahlen sind: Ist `quantity` ein `int`, `unit_price` ein `float`, `discount` ein `int`? Wandle dann die Texte passend um und berechne den Endbetrag: Menge mal Stückpreis minus Rabatt. Erwartet: `54.97`.

````python
quantity = "3"  # aus dem Formular
unit_price = "19.99"  # aus dem Formular
discount = 5  # von der Kasse
````

## Zusammenfassung

- In Python ist **alles ein Objekt**: Zahlen, Strings, Funktionen, Typen.
- Operatoren und eingebaute Funktionen rufen **Dunder-Methoden** auf: `1 + 2` ist `(1).__add__(2)`, `len(x)` ist `x.__len__()`.
- Variablen sind Namen ohne Typ, der Typ gehört zum Wert.
- Eingebaute Funktionen sind überschreibbare Namen. Also keine Variable `list`, `sum` oder `max` nennen. Schlüsselwörter wie `for` lassen sich nicht überschreiben.
- `type()`, `isinstance()`, `dir()` und `help()` helfen beim Erkunden.
