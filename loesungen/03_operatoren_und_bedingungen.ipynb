## Bedingungen mit `if`

Der eingerückte Block läuft nur, wenn die Bedingung wahr ist. Einrückung: **4 Leerzeichen**.

````python
temperature = 28

if temperature > 25:
    print("Ab ins Freibad!")
    print("Sonnencreme nicht vergessen.")
print("Diese Zeile läuft immer.")
````

Mit `elif` und `else` prüfst du mehrere Fälle. Der erste zutreffende gewinnt.

````python
temperature = 12

if temperature > 25:
    print("heiß")
elif temperature > 15:
    print("angenehm")
elif temperature > 5:
    print("frisch")
else:
    print("kalt")
````

Schreibe die Abfrage mit `if`/`elif`/`else` und prüfe sie mit 95, 75, 60 und 30 Punkten: Ändere `points` und führe die Zelle jeweils erneut aus.

Untersuchen: Was geht schief, wenn du `points >= 50` als erste Bedingung prüfst?

````python
points = 95  # auch 75, 60 und 30 probieren

if points >= 90:
    print("sehr gut")
elif points >= 75:
    print("gut")
elif points >= 50:
    print("bestanden")
else:
    print("nicht bestanden")
# 95 -> sehr gut, 75 -> gut, 60 -> bestanden, 30 -> nicht bestanden

# Untersuchen: falsche Reihenfolge
if points >= 50:
    print("bestanden")
elif points >= 75:
    print("gut")
elif points >= 90:
    print("sehr gut")
else:
    print("nicht bestanden")
# 95 -> bestanden: Der erste zutreffende Zweig gewinnt, und ab 50 trifft
# schon die erste Bedingung zu. "gut" und "sehr gut" erreicht niemand.
````

Für einfache Fälle gibt es den **bedingten Ausdruck** (im Buch: Ternary-Operator). Er liefert einen von zwei Werten und passt in eine Zeile:

```
wert_wenn_wahr if bedingung else wert_wenn_falsch
```

````python
age = 16
status = "volljährig" if age >= 18 else "minderjährig"
status
````

> Nimm den bedingten Ausdruck nur, wenn er kurz und leicht lesbar bleibt. Verschachtelt wie `a if x else b if y else c` versteht ihn niemand mehr; dann lieber `if`/`elif`.

**Übung 3 (Versandkosten, leicht):** Ab 50 € Bestellwert ist der Versand kostenlos, sonst kostet er 4,95 €. Berechne `shipping` zuerst mit einem `if`/`else`-Block, dann in einer Zeile mit dem bedingten Ausdruck, und gib jeweils den Gesamtbetrag aus. Prüfe auch genau 50 €. Welche Fassung nimmst du, und warum?

````python
order_total = 42.80  # auch 50 probieren

# mit if/else-Block
if order_total >= 50:
    shipping = 0
else:
    shipping = 4.95
print(f"Gesamt: {order_total + shipping:.2f} €")

# in einer Zeile mit dem bedingten Ausdruck
shipping = 0 if order_total >= 50 else 4.95
print(f"Gesamt: {order_total + shipping:.2f} €")

# 42.80 -> Gesamt: 47.75 €, 50 -> Gesamt: 50.00 €
# Hier passt die eine Zeile: Es wird nur ein Wert gewählt. Sobald ein Zweig
# mehr tun soll als einen Wert liefern, ist der Block lesbarer.
````

## Fallunterscheidung mit `match`

Vergleichst du **einen Wert** mit mehreren Möglichkeiten, ist `match` übersichtlicher als eine lange `elif`-Kette. Python prüft die `case`-Zweige von oben nach unten, der erste passende gewinnt.

````python
command = "halt"

match command:
    case "start":
        print("Motor läuft")
    case "stop" | "halt":
        print("Motor aus")
    case _:
        print(f"Unbekannter Befehl: {command}")
````

- `|` bedeutet „oder“: Der Zweig passt auf `"stop"` und auf `"halt"`.
- `case _:` passt auf alles, was übrig bleibt, wie `else`.
- Für Java- und C#-Umsteiger: `match` braucht kein `break`. Es läuft immer genau ein Zweig, danach geht es hinter dem `match` weiter.

### Zusatzbedingungen: Guards

Hinter ein Muster darfst du `if` schreiben. Der Zweig passt nur, wenn auch diese Bedingung wahr ist. `case t if …` bindet den Wert dabei an den Namen `t`:

````python
temperature = 12

match temperature:
    case t if t > 25:
        print("heiß")
    case t if t > 15:
        print("angenehm")
    case _:
        print("frisch bis kalt")
````

Für reine Größer-/Kleiner-Vergleiche bleibt `if`/`elif` meist lesbarer. Seine Stärke zeigt `match`, wenn ein Wert eine **Struktur** hat.

### Muster für Listen

`match` kann eine Liste zerlegen und die Teile direkt an Namen binden. Typisches Beispiel: ein Befehl mit Argumenten, wie ihn ein Spiel oder ein Kommandozeilen-Tool einliest. Probier verschiedene Eingaben aus:

````python
user_input = "move 3 4"

match user_input.split():
    case ["quit"]:
        print("Tschüss")
    case ["move", x, y]:
        print(f"Gehe nach ({x}, {y})")
    case ["say", *words]:
        print(" ".join(words))
    case _:
        print("Unbekannter Befehl")
````

`*words` sammelt alle restlichen Elemente ein, wie beim Entpacken. Es gibt auch Muster für Dictionaries und Objekte; `match` zusammen mit Aufzählungen (`Enum`) zeigt Notebook 12.

> **Typischer Fehler:** Ein nackter Name im `case` vergleicht **nicht**, er passt auf alles und bindet den Wert an diesen Namen. `case limit:` trifft also immer zu, auch wenn es eine Variable `limit` gibt. Vergleiche deshalb mit Literalen (`case "stop":`) oder mit Namen, die einen Punkt enthalten (`case Color.RED:`, siehe Notebook 12).

**Übung 4 (Ja/Nein-Abfrage, leicht):** Code, den eine KI geschrieben haben könnte (präpariert, ein Fehler eingebaut): Er soll nur bei `j` oder `ja` „Wird gelöscht“ ausgeben, tut es aber bei jeder Antwort. Finde den Fehler und behebe ihn einmal mit `or` und einmal mit `match`.

## None und Vergleiche: `==` oder `is`?

`None` kennst du aus Notebook 01 als „kein Wert“. Du brauchst es, wenn eine Angabe fehlt oder noch unbekannt ist:

- eine optionale Angabe, z. B. kein Rabattcode eingegeben,
- ein Platzhalter, der erst später gefüllt wird.

Auch Funktionen ohne `return` liefern `None`; `print("Hallo")` zum Beispiel gibt Text aus und liefert `None` zurück. Mehr dazu in Notebook 07 (Funktionen).

````python
discount_code = None  # der Kunde hat keinen Rabattcode eingegeben

if discount_code is None:
    print("Bestellung ohne Rabattcode")
````

### `==` oder `is`?

- `==` vergleicht **Werte**: Haben beide denselben Inhalt?
- `is` prüft **Identität**: Ist es dasselbe Objekt? (Details in Notebook 06.)

Die Regel (PEP 8): Werte vergleichst du immer mit `==`. `is` und `is not` nimmst du nur für `None`. `True` und `False` vergleichst du gar nicht, weder mit `== True` noch mit `is True`, sondern schreibst `if flag:` bzw. `if not flag:`.

````python
quantity = 3
print(quantity == 3)  # Werte: ==

discount_code = None
print(discount_code is None)  # None: is
print(discount_code is not None)

is_member = True
if is_member:  # nicht: if is_member == True
    print("Mitgliederpreis")
````

Vergleichst du Zahlen oder Texte mit `is`, ist das ein Fehler, auch wenn er manchmal das richtige Ergebnis liefert:

````python
x = int("1000")  # z. B. aus einer Eingabe
print(x == 1000, x is 1000)

name = " Ada ".strip()  # z. B. aus einem Formular
print(name == "Ada", name is "Ada")
````

Python warnt schon beim Lesen der Zelle mit einer `SyntaxWarning` („Did you mean "=="?“). `==` stimmt in beiden Fällen, `is` einmal ja, einmal nein: Ob zwei gleiche Zahlen oder Texte dasselbe Objekt sind, ist Zufall der Implementierung. Mehr zu Identität in Notebook 06.

### Falle: `None` ist falsy

Wie `0`, `""` und `[]` gilt `None` in einer Bedingung als falsch (Notebook 01: `bool(None)` ist `False`). `if not x:` trifft deshalb auch zu, wenn `x` gültig `0` ist. Meinst du „nicht angegeben“, prüfe mit `is None`:

````python
temperature = 0  # 0 °C ist ein gültiger Messwert, None hieße: Sensor hat nichts geliefert

if not temperature:
    print("if not:     kein Messwert")  # falsch, es gibt einen Messwert

if temperature is None:
    print("if is None: kein Messwert")
else:
    print(f"if is None: {temperature} °C")
````

`None` ist mit nichts gleich außer mit sich selbst; `None == 0` ist `False`. Größer- und Kleiner-Vergleiche mit `None` sind gar nicht erlaubt und werfen einen `TypeError`:

````python
print(None == 0)
None < 1
````

**Übung 8 (Smart-Home-Befehl, schwer):** Der Startcode versteht die Befehle `"licht an"` und `"licht aus"`. Sage vorher, was er bei `"licht an"`, `"Licht an"` und `"heizung 21"` ausgibt. Erweitere ihn: Fasse beide Lichtbefehle in einem `case` mit `|` und `as` zusammen und ergänze den Befehl `"heizung <Grad>"`, der nur 5 bis 30 Grad annimmt. Was passiert, wenn du im Guard `int()` weglässt?

- **Untersuchen:** Warum ist `"Licht an"` ungültig? Was liefert `command.split()`?
- **Ändern:** Fasse die Lichtbefehle zusammen und ergänze die Heizung mit einem Guard.

````python
command = "heizung 21"
match command.split():
    case ["licht", "an"]:
        print("Licht an")
    case ["licht", "aus"]:
        print("Licht aus")
    case _:
        print("ungültig")
````

````python
command = "heizung 21"
match command.split():
    case ["licht", "an" | "aus" as state]:
        print(f"Licht {state}")
    case ["heizung", value] if 5 <= int(value) <= 30:  # ohne int(): TypeError
        print(f"Heizung auf {value} °C")
    case _:
        print("ungültig")
````

## Zusammenfassung

- `//` und `%` liefern Ganzzahl-Ergebnis und Rest.
- Vergleiche liefern `bool` und lassen sich verketten: `0 < x < 100`.
- `and`, `or`, `not` verknüpfen Bedingungen.
- `if` / `elif` / `else` mit 4 Leerzeichen Einrückung.
- Bedingter Ausdruck: `a if bedingung else b`, nur für kurze Fälle.
- `match` / `case` vergleicht einen Wert mit Mustern: `|` für „oder“, `case _` für den Rest, `if` als Guard, Listenmuster wie `["move", x, y]`.
- `input()` liefert immer `str`, also umwandeln.
- Werte mit `==` vergleichen, `None` mit `is None`; `if not x:` trifft auch bei `0` und `""` zu.
