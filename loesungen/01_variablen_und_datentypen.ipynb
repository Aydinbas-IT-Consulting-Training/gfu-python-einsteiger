## Datentypen

Jeder **Wert** hat einen Typ. `type()` verrät ihn:

````python
print(type(42))
print(type(3.1415))
print(type("Hallo"))
print(type(True))
print(type(None))
````

| Typ | Bedeutung | Beispiele |
| --- | --- | --- |
| `int` | ganze Zahl, beliebig groß | `42`, `-7`, `1_000_000` |
| `float` | Kommazahl | `3.14`, `-0.5`, `1e6` |
| `str` | Text | `"Hallo"`, `'Welt'` |
| `bool` | Wahrheitswert | `True`, `False` |
| `NoneType` | „kein Wert“ | `None` |

### Vorsicht bei Kommazahlen

`float` speichert Zahlen binär und damit nicht immer exakt:

````python
0.1 + 0.2
````

Für Geldbeträge gibt es deshalb das Modul `decimal`. Wichtig: Die Zahl als **String** übergeben.

````python
from decimal import Decimal

Decimal("0.1") + Decimal("0.2")
````

Für Vergleiche von Kommazahlen nutzt man `math.isclose`:

````python
import math

math.isclose(0.1 + 0.2, 0.3)
````

## Typumwandlung

Mit `int()`, `float()`, `str()` und `bool()` wandelst du Werte um.

````python
print(int("42") + 1)
print(float("3.5") * 2)
print(str(42) + " ist die Antwort")
print(int(3.99))  # schneidet ab, rundet nicht
print(round(3.99))
````

Python wandelt **nicht** automatisch zwischen Text und Zahl um. *Explicit is better than implicit:* Du entscheidest, was gemeint ist.

````python
"Alter: " + 18
````

````python
int("drei")
````

Jeder Wert hat außerdem einen Wahrheitswert. „Leeres“ ist `False`:

````python
print(bool(0), bool(""), bool(None))
print(bool(7), bool("Hallo"))
````

## Formatierte Ausgabe mit f-Strings

Ein f-String beginnt mit `f`. In geschweiften Klammern steht beliebiger Python-Code:

````python
name = "Michael"
age = 18
print(f"Hallo {name}, du bist {age} Jahre alt.")
print(f"Nächstes Jahr bist du {age + 1}.")
````

Nach einem Doppelpunkt folgt das **Format**:

````python
price = 1234.5678
print(f"Preis: {price:.2f} €")        # 2 Nachkommastellen
print(f"Preis: {price:,.2f} €")       # mit Tausendertrennzeichen
print(f"Anteil: {0.256:.1%}")         # Prozent
print(f"|{'links':<10}|{'rechts':>10}|")  # Ausrichtung
````

Zum Debuggen praktisch: `=` zeigt Name und Wert.

````python
print(f"{age=}")
````

> Ältere Schreibweisen wie `"%s" % name` oder `"{}".format(name)` begegnen dir in altem Code. Für neuen Code nimmst du f-Strings.

## Übungen

**Übung 1 (Rezept umrechnen, leicht):** Ein Rezept für 4 Personen braucht 250 g Mehl und 3 Eier. Rechne auf 6 Personen um: Mehl auf ganze Gramm, Eier auf ganze Stück.

````python
factor = 6 / 4
print("Mehl:", round(250 * factor), "g")
print("Eier:", round(3 * factor))
# Hinweis: 3 * 1.5 = 4.5 Eier ergibt 4, nicht 5. round() rundet genau .5
# zur geraden Zahl (Banker's Rounding, "round half to even").
````

**Übung 2 (Zahlen aus einem Formular, leicht):** Ein Bestellformular liefert Preis und Menge als Text, so wie sie auch `input()` liefert: `price = "4.99"` und `quantity = "3"`. Welche Ausgabe erzeugt `price + quantity`, und warum? Warum bricht `price * quantity` mit einem `TypeError` ab? Die zweite Zelle darunter läuft absichtlich in diesen Fehler.

Wandle dann beide Werte passend um, berechne den Gesamtpreis und gib ihn mit einem f-String und zwei Nachkommastellen aus: `Gesamt: 14.97 €`. Löse es einmal mit `float()` und einmal mit `Decimal`. Welche Fassung nimmst du, und warum?

````python
price = "4.99"  # aus dem Formular
quantity = "3"
print(price + quantity)
````

````python
price * quantity
````

````python
from decimal import Decimal

price = "4.99"
quantity = "3"
print(price + quantity)  # 4.993: + hängt zwei Texte aneinander
# price * quantity -> TypeError: Text mal Text ist nicht definiert

# mit float()
total = float(price) * int(quantity)
print(f"Gesamt: {total:.2f} €")  # Gesamt: 14.97 €

# mit Decimal: der Text aus dem Formular geht direkt hinein, das Ergebnis ist exakt
total = Decimal(price) * int(quantity)
print(f"Gesamt: {total:.2f} €")  # Gesamt: 14.97 €

# Für Geldbeträge passt Decimal. float liefert hier zufällig genau 14.97,
# bei anderen Beträgen nicht (denk an 0.1 + 0.2).
````

**Übung 3 (Einkaufsliste als Tabelle, leicht):** Der Startcode enthält deinen Einkauf: Menge und Stückpreis für Milch, Brot und Eier. Gib ihn mit f-Strings als Tabelle aus: Artikel linksbündig in 12 Zeichen, Menge rechtsbündig in 4 Zeichen, Zeilenpreis (Menge mal Stückpreis) rechtsbündig in 8 Zeichen mit zwei Nachkommastellen. Zum Schluss folgt eine Zeile mit der Summe:

```
Milch          2    2.18 €
Brot           1    3.49 €
Eier           6    1.92 €
Summe               7.59 €
```

````python
from decimal import Decimal

milk_quantity, milk_price = 2, Decimal("1.09")
bread_quantity, bread_price = 1, Decimal("3.49")
egg_quantity, egg_price = 6, Decimal("0.32")
````

````python
milk_total = milk_quantity * milk_price
bread_total = bread_quantity * bread_price
egg_total = egg_quantity * egg_price
total = milk_total + bread_total + egg_total

print(f"{'Milch':<12}{milk_quantity:>4}{milk_total:>8.2f} €")
print(f"{'Brot':<12}{bread_quantity:>4}{bread_total:>8.2f} €")
print(f"{'Eier':<12}{egg_quantity:>4}{egg_total:>8.2f} €")
print(f"{'Summe':<16}{total:>8.2f} €")  # 16 = 12 + 4: die Mengenspalte bleibt leer
````

**Übung 4 (Kassenbon mit Decimal, mittel):** Der Startcode rechnet deinen Einkauf (3 Brötchen zu 0,35 €, ein Kaffee zu 2,90 €) mit `float`. Sage die Ausgabe vorher, führe den Code dann aus und erweitere ihn: Rechne mit `Decimal` und gib Summe und enthaltene Mehrwertsteuer (7 %) rechtsbündig mit zwei Nachkommastellen aus.

- **Untersuchen:** Woher kommt `3.9499999999999997`? Was zeigt `Decimal(0.35)` ohne Anführungszeichen?
- **Ändern:** Stelle die Rechnung auf `Decimal` um, rechne die MwSt. heraus und richte die Ausgabe aus.

````python
roll, coffee = 0.35, 2.90
total = 3 * roll + coffee
print("Summe:", total, "€")
````

````python
from decimal import Decimal

roll, coffee = Decimal("0.35"), Decimal("2.90")
total = 3 * roll + coffee
vat = total - total / Decimal("1.07")
print(f"Summe:      {total:>6.2f} €")
print(f"darin MwSt: {vat:>6.2f} €")
print(Decimal(0.35))  # 0.34999999999999997779553950749686919152736663818359375
# 3.9499999999999997: float speichert 0.35 binär und damit nur ungefähr.
# Decimal(0.35) übernimmt genau diesen ungenauen float-Wert, deshalb Decimal("0.35").
````

**Übung 5 (Lauf-Pace aus App-Daten, mittel):** Deine Lauf-App liefert Text: `distance = "10.5"` (km) und `duration = "58"` (Minuten). Der Startcode gibt die Pace als Kommazahl aus. Sage die Ausgabe vorher, führe den Code dann aus und bringe die Ausgabe ins übliche Format `5:31 min/km`: erst auf ganze Sekunden runden, dann in Minuten und Sekunden zerlegen.

- **Untersuchen:** Bedeutet 5.52 "5 Minuten 52 Sekunden"? Was passiert, wenn du `float()` weglässt?
- **Ändern:** Gib die Pace im Format `5:31 min/km` aus.

````python
distance = "10.5"  # km
duration = "58"  # Minuten
pace = float(duration) / float(distance)
print(f"Pace: {pace:.2f} min/km")
````

````python
distance = "10.5"  # km, aus der App
duration = "58"  # Minuten
pace = float(duration) / float(distance)  # Minuten pro km
total_seconds = round(pace * 60)
minutes = int(total_seconds / 60)
seconds = total_seconds - minutes * 60
print(f"Pace: {minutes}:{seconds:02d} min/km")
# 5.52 heißt 5,52 Minuten: 5 Minuten und gut 0,5 Minuten, also rund 31 Sekunden.
# Ohne float(): "58" / "10.5" -> TypeError, mit Text lässt sich nicht rechnen.
````

**Übung 6 (Jahresabrechnung Strom, schwer):** Der Startcode rechnet aus zwei abgelesenen Zählerständen die Arbeitskosten. Sage die Ausgabe vorher, führe den Code dann aus und erweitere ihn zu einer Jahresabrechnung: 12,50 € Grundpreis pro Monat, 12 gezahlte Abschläge zu 85 €, Differenz mit Vorzeichen (plus heißt Nachzahlung), Beträge rechtsbündig.

- **Untersuchen:** Was passiert, wenn du die beiden `int()` weglässt? Warum ist der Strompreis eine Konstante?
- **Ändern:** Ergänze Grundpreis, Abschläge und Differenz und gib alles als ausgerichtete Abrechnung aus.

````python
from decimal import Decimal

old_reading, new_reading = "48213", "51007"  # kWh, vom Zähler abgelesen
PRICE_PER_KWH = Decimal("0.32")
consumption = int(new_reading) - int(old_reading)
print("Verbrauch:", consumption, "kWh, Kosten:", consumption * PRICE_PER_KWH, "€")
````

````python
from decimal import Decimal

old_reading, new_reading = "48213", "51007"  # kWh, vom Zähler abgelesen
PRICE_PER_KWH = Decimal("0.32")
BASE_PRICE_PER_MONTH = Decimal("12.50")
consumption = int(new_reading) - int(old_reading)
cost = consumption * PRICE_PER_KWH + 12 * BASE_PRICE_PER_MONTH
paid = 12 * Decimal("85")
print(f"Verbrauch: {consumption:>9} kWh")
print(f"Kosten:    {cost:>9.2f} €")
print(f"Abschläge: {paid:>9.2f} €")
print(f"Differenz: {cost - paid:>+9.2f} €  (plus = Nachzahlung)")
# Ohne int(): "51007" - "48213" -> TypeError, Text lässt sich nicht subtrahieren.
# Konstante: Der Preis ändert sich nur mit dem Tarif, nicht im Programm.
````

## Zusammenfassung

- Variablen entstehen durch Zuweisung: `name = wert`.
- Grundtypen: `int`, `float`, `str`, `bool`, `None`.
- Umwandeln mit `int()`, `float()`, `str()`, `bool()`.
- `float` ist ungenau; für Geld `Decimal("…")`, für Vergleiche `math.isclose`.
- Ausgaben mit f-Strings: `f"{wert:.2f}"`.
